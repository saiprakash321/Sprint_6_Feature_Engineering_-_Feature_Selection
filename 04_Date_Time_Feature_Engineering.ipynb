## 1. Year, Month, Day

### Explanation

Year, Month, and Day features split a date into separate numerical components.

In the Hotel Booking dataset, the arrival date is represented using:

- `arrival_date_year`
- `arrival_date_month`
- `arrival_date_day_of_month`

These values can be used to create a complete arrival date and extract its individual components.

### Example

Arrival Date = 2017-07-15

→ Year = 2017  
→ Month = 7  
→ Day = 15

### Why we use / need

Machine Learning models cannot directly understand a raw date as a meaningful numerical feature.

Breaking a date into Year, Month, and Day allows the model to identify yearly, monthly, and daily patterns.

### When to use

Use these features when analyzing:

- Yearly booking trends
- Monthly booking patterns
- Daily booking behavior
- Seasonal changes in hotel demand

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# Create a complete arrival date

df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str)
    + "-"
    + df["arrival_date_month"]
    + "-"
    + df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d"
)

# Extract Year, Month, and Day

df["Year"] = df["arrival_date"].dt.year
df["Month"] = df["arrival_date"].dt.month
df["Day"] = df["arrival_date"].dt.day

df[
    [
        "arrival_date",
        "Year",
        "Month",
        "Day"
    ]
].head()
```

### Observation

The arrival date has been converted into a datetime column and separated into Year, Month, and Day.

These features make it easier to analyze booking patterns across different years, months, and days.

## 2. Day of Week

### Explanation

Day of Week identifies which day of the week the hotel booking arrival occurred.

The `arrival_date` column can be used to extract both the numerical day of the week and its name.

### Example

If the arrival date is a Monday:

→ DayOfWeek = 0  
→ DayName = Monday

### Why we use / need

Booking behavior can differ across weekdays and weekends.

This feature helps the model identify weekly booking patterns and differences in hotel demand.

### When to use

Use this feature when analyzing:

- Weekly booking patterns
- Weekday vs weekend demand
- Hotel occupancy trends
- Customer booking behavior

```python
df["DayOfWeek"] = df["arrival_date"].dt.dayofweek
df["DayName"] = df["arrival_date"].dt.day_name()

df[
    [
        "arrival_date",
        "DayOfWeek",
        "DayName"
    ]
].head()
```

### Observation

The day of the week has been extracted from the arrival date.

`DayOfWeek` represents Monday as 0 and Sunday as 6, while `DayName` provides the corresponding day name.

This feature can help identify weekly patterns in hotel bookings.

## 3. Week Number

### Explanation

Week Number shows which week of the year the hotel booking arrival date falls into.

The week number ranges from 1 to 53 depending on the year.

### Example

An arrival date in the first week of the year:

→ WeekNumber = 1

### Why we use / need

Weekly grouping can reduce daily-level noise while still capturing short-term booking patterns.

This feature helps the model identify recurring weekly trends in hotel bookings.

### When to use

Use this feature for:

- Weekly booking analysis
- Demand forecasting
- Occupancy trend analysis
- Seasonal booking patterns
- Campaign or promotion analysis

```python
df["WeekNumber"] = (
    df["arrival_date"]
    .dt.isocalendar()
    .week
)

df[
    [
        "arrival_date",
        "WeekNumber"
    ]
].head()
```

### Observation

The ISO week number has been extracted from the arrival date.

The `WeekNumber` feature indicates the week of the year in which each hotel booking arrival occurred.

This can help identify weekly booking and demand patterns.

## 5. Weekend Indicator

### Explanation

Weekend Indicator is a binary feature that shows whether the hotel booking arrival occurred on a weekend.

It represents:

- 1 – Weekend
- 0 – Weekday

### Example

Saturday arrival → IsWeekend = 1

Monday arrival → IsWeekend = 0

### Why we use / need

Hotel booking behavior can differ between weekdays and weekends.

A direct weekend indicator helps the Machine Learning model identify these differences and learn weekly booking patterns.

### When to use

Use this feature when:

- Weekend demand is different from weekday demand.
- Analyzing hotel occupancy patterns.
- Studying customer booking behavior.
- Building hotel demand or cancellation prediction models.

```python
df["IsWeekend"] = (
    df["DayOfWeek"]
    .isin([5, 6])
    .astype(int)
)

df[
    [
        "arrival_date",
        "DayOfWeek",
        "DayName",
        "IsWeekend"
    ]
].head()
```

### Observation

The `IsWeekend` feature has been created using the day of the week.

Saturday and Sunday are represented as 1, while weekdays are represented as 0.

This binary feature makes it easier for a Machine Learning model to identify differences between weekend and weekday hotel bookings.

## 6. Month Start / Month End

### Explanation

Month Start and Month End features indicate whether the hotel booking arrival date falls on the first or last day of a month.

They are represented as binary features:

- `IsMonthStart = 1` → First day of the month
- `IsMonthEnd = 1` → Last day of the month

### Example

July 1 → IsMonthStart = 1

July 31 → IsMonthEnd = 1

### Why we use / need

Booking demand can vary around the beginning and end of a month.

These features allow the model to capture possible booking patterns related to month boundaries.

### When to use

Use these features when analyzing:

- Monthly booking patterns
- Hotel demand
- Occupancy trends
- Month-based customer behavior

```python
df["IsMonthStart"] = (
    df["arrival_date"]
    .dt.is_month_start
    .astype(int)
)

df["IsMonthEnd"] = (
    df["arrival_date"]
    .dt.is_month_end
    .astype(int)
)

df[
    [
        "arrival_date",
        "IsMonthStart",
        "IsMonthEnd"
    ]
].head()
```

### Observation

The `IsMonthStart` and `IsMonthEnd` features have been created from the arrival date.

They contain 1 when the booking arrival occurs on the first or last day of the month and 0 otherwise.

These features can help the model identify patterns occurring around monthly boundaries.

## 7. Days Since Event

### Explanation

Days Since Event measures how many days have passed between a fixed reference date and the hotel booking arrival date.

For the Hotel Booking dataset, the earliest arrival date can be used as the reference event.

### Example

First arrival date = January 1  
Booking arrival date = January 10

→ DaysSinceFirstEvent = 9 days

### Why we use / need

A raw date does not directly represent how much time has passed.

Converting the date into the number of elapsed days provides a continuous numerical feature that Machine Learning models can use more easily.

### When to use

Use this feature when tracking:

- Time since the first booking
- Time since a campaign or event
- Booking timeline progression
- Long-term booking trends

```python
# Find the earliest arrival date
first_date = df["arrival_date"].min()

# Calculate days since the first event
df["DaysSinceFirstEvent"] = (
    df["arrival_date"] - first_date
).dt.days

df[
    [
        "arrival_date",
        "DaysSinceFirstEvent"
    ]
].head()
```

### Observation

The number of days from the earliest arrival date has been calculated for each booking.

The first arrival date has a value of 0, while later arrival dates have increasing values.

This converts the date into a continuous time-based feature that can help the model understand the progression of bookings over time.

## 8. Time Difference

### Explanation

Time Difference calculates the gap between consecutive hotel booking arrivals for the same customer.

In the Hotel Booking dataset, `arrival_date` and `customer_type` are available, but there is no `CustomerID` column.

Therefore, a customer-level time difference cannot be calculated directly from the dataset.

For this dataset, we can calculate the difference between consecutive arrival dates after sorting the bookings by date.

### Example

Arrival 1 = January 10  
Arrival 2 = January 15

→ TimeDifference = 5 days

### Why we use / need

Time difference converts consecutive dates into a numerical measure of elapsed time.

It can help identify booking intervals and time-based patterns in the data.

### When to use

Use this feature when:

- Consecutive events need to be compared.
- Booking intervals are important.
- Time-based behavior needs to be analyzed.
- Customer-level transaction history is available.

```python
# Sort bookings by arrival date
df = df.sort_values("arrival_date").copy()

# Calculate the difference between consecutive arrival dates
df["TimeDifferenceDays"] = (
    df["arrival_date"].diff().dt.days
)

df[
    [
        "arrival_date",
        "TimeDifferenceDays"
    ]
].head(10)
```

### Observation

The `TimeDifferenceDays` feature calculates the number of days between consecutive booking arrival dates.

The first row has a missing value because there is no previous arrival date for comparison.

Since the Hotel Booking dataset does not contain a customer identifier, this calculation represents the overall sequence of arrival dates rather than the time difference between purchases for an individual customer.

## 13. Seasonality

### Explanation

Seasonality represents recurring patterns that occur during specific periods of the year.

In the Hotel Booking dataset, booking demand can vary across different months and seasons.

The month of the arrival date can be mapped into broader seasons:

- Winter – December, January, February
- Spring – March, April, May
- Summer – June, July, August
- Autumn – September, October, November

### Example

December → Winter

July → Summer

### Why we use / need

Hotel bookings and demand can change depending on the time of year.

Creating a season feature helps the Machine Learning model capture these recurring seasonal patterns.

### When to use

Use Seasonality for:

- Hotel demand forecasting
- Booking pattern analysis
- Occupancy prediction
- Seasonal pricing analysis
- Tourism-related business analysis

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"


df["Season"] = df["Month"].apply(get_season)

df[
    [
        "arrival_date",
        "Month",
        "Season"
    ]
].head()
```

### Observation

The `Month` feature has been converted into a seasonal category.

Each booking is assigned to Winter, Spring, Summer, or Autumn based on its arrival month.

This feature helps capture recurring seasonal patterns in hotel bookings and can provide useful information for Machine Learning models.

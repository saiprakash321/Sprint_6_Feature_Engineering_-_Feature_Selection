## 4. Create Numerical Features

Numerical feature engineering creates new numerical features by combining or transforming existing numerical columns.

For the Hotel Bookings dataset, we can create useful numerical features such as:

- Total_Stay_Nights
- Total_Guests
- Estimated_Stay_Revenue
- Guests_Per_Night
- ADR_Per_Guest
- Special_Request_Rate

These features combine existing booking information to provide additional numerical information for machine learning.

```python
df["children"] = df["children"].fillna(0)

df["Total_Stay_Nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["Total_Guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

df["Estimated_Stay_Revenue"] = (
    df["adr"] * df["Total_Stay_Nights"]
)

df["Guests_Per_Night"] = np.where(
    df["Total_Stay_Nights"] > 0,
    df["Total_Guests"] / df["Total_Stay_Nights"],
    0
)

df["ADR_Per_Guest"] = np.where(
    df["Total_Guests"] > 0,
    df["adr"] / df["Total_Guests"],
    0
)

df["Special_Request_Rate"] = np.where(
    df["Total_Stay_Nights"] > 0,
    df["total_of_special_requests"] / df["Total_Stay_Nights"],
    0
)

df[
    [
        "Total_Stay_Nights",
        "Total_Guests",
        "Estimated_Stay_Revenue",
        "Guests_Per_Night",
        "ADR_Per_Guest",
        "Special_Request_Rate"
    ]
].head()
```

### Observation

Several new numerical features have been created from the existing Hotel Bookings data.

These features provide additional information about stay duration, guest count, booking value, pricing, and special-request behavior.

### Code Explanation

- `Total_Stay_Nights` combines weekend and weekday nights.
- `Total_Guests` combines adults, children, and babies.
- `Estimated_Stay_Revenue` combines ADR with total stay duration.
- `Guests_Per_Night` represents guests relative to stay duration.
- `ADR_Per_Guest` represents ADR relative to guest count.
- `Special_Request_Rate` represents special requests relative to stay duration.
- `np.where()` is used to safely handle zero values during division.

These engineered numerical features will be considered in the later feature selection stage.

## 5. Create Categorical Features

Categorical feature engineering converts existing information into meaningful categories that can be used by machine learning models.

For the Hotel Bookings dataset, we can create categorical features such as:

- Booking_Lead_Category
- Guest_Composition
- Booking_Value_Category
- Seasonal_Booking

These features convert numerical or date-related information into meaningful business categories.

Categorical features can help a model identify different booking patterns and guest groups.

```python
# Booking lead-time category
df["Booking_Lead_Category"] = pd.cut(
    df["lead_time"],
    bins=[-1, 30, 90, np.inf],
    labels=["Short Lead", "Medium Lead", "Long Lead"]
)

# Guest composition
def get_guest_composition(row):
    if row["children"] > 0 or row["babies"] > 0:
        return "Family"
    elif row["adults"] == 1:
        return "Solo"
    else:
        return "Adults Only"

df["Guest_Composition"] = df.apply(
    get_guest_composition,
    axis=1
)

# Booking value category
df["Booking_Value_Category"] = pd.cut(
    df["Estimated_Stay_Revenue"],
    bins=[-np.inf, 500, 1500, np.inf],
    labels=["Low Value", "Medium Value", "High Value"]
)

# Seasonal booking
season_map = {
    "December": "Winter",
    "January": "Winter",
    "February": "Winter",
    "March": "Spring",
    "April": "Spring",
    "May": "Spring",
    "June": "Summer",
    "July": "Summer",
    "August": "Summer",
    "September": "Autumn",
    "October": "Autumn",
    "November": "Autumn"
}

df["Seasonal_Booking"] = df["arrival_date_month"].map(season_map)

df[
    [
        "Booking_Lead_Category",
        "Guest_Composition",
        "Booking_Value_Category",
        "Seasonal_Booking"
    ]
].head()
```

### Observation

Several categorical features have been created from the existing numerical and date-related information.

These features represent different booking behaviors, guest types, booking values, and seasonal patterns.

### Code Explanation

- `Booking_Lead_Category` groups bookings based on how early they were made.
- `Guest_Composition` identifies Family, Solo, and Adults Only bookings.
- `Booking_Value_Category` groups bookings based on estimated stay revenue.
- `Seasonal_Booking` converts the arrival month into a seasonal category.
- These categorical features provide additional business meaning and will be evaluated during feature selection.

## 6. Create Date/Time Features

Date and time information can contain useful patterns related to hotel bookings.

The Hotel Bookings dataset stores arrival information in separate columns:

- arrival_date_year
- arrival_date_month
- arrival_date_day_of_month

These columns can be combined to create a proper arrival date and additional date-based features.

For this workflow, we will create:

- Arrival_Date
- Arrival_Year
- Arrival_Month
- Arrival_Day
- Arrival_DayOfWeek
- Is_Weekend_Arrival

```python
month_map = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["Arrival_Month_Number"] = df["arrival_date_month"].map(month_map)

df["Arrival_Date"] = pd.to_datetime(
    dict(
        year=df["arrival_date_year"],
        month=df["Arrival_Month_Number"],
        day=df["arrival_date_day_of_month"]
    ),
    errors="coerce"
)

df["Arrival_Year"] = df["Arrival_Date"].dt.year
df["Arrival_Month"] = df["Arrival_Date"].dt.month
df["Arrival_Day"] = df["Arrival_Date"].dt.day
df["Arrival_DayOfWeek"] = df["Arrival_Date"].dt.dayofweek
df["Is_Weekend_Arrival"] = (
    df["Arrival_DayOfWeek"] >= 5
).astype(int)

df[
    [
        "Arrival_Date",
        "Arrival_Year",
        "Arrival_Month",
        "Arrival_Day",
        "Arrival_DayOfWeek",
        "Is_Weekend_Arrival"
    ]
].head()
```

### Observation

The separate arrival-date columns have been combined to create a proper `Arrival_Date`.

Additional date-based features have also been created to represent the year, month, day, day of the week, and weekend arrival information.

### Code Explanation

- `month_map` converts month names into numerical month values.
- `pd.to_datetime()` combines the year, month, and day into `Arrival_Date`.
- `.dt.year`, `.dt.month`, and `.dt.day` extract date components.
- `.dt.dayofweek` identifies the day of the week.
- `Is_Weekend_Arrival` indicates whether the arrival falls on Saturday or Sunday.
- These features can help capture seasonal and timing-related booking patterns.

## 7. Create Aggregation Features

Aggregation features summarize information from multiple records or related groups.

In the Hotel Bookings dataset, group-based aggregations can provide additional information about booking patterns for each hotel.

For example, we can calculate:

- Booking count
- Average ADR
- Total stay nights

These aggregated values can provide useful context about hotel-level booking behavior.

```python
hotel_aggregations = df.groupby("hotel").agg(
    Booking_Count=("hotel", "count"),
    Average_ADR=("adr", "mean"),
    Total_Stay_Nights=("Total_Stay_Nights", "sum")
).reset_index()

hotel_aggregations
```

### Observation

Hotel-level aggregation features have been created to summarize booking behavior for each hotel type.

The aggregated information includes the number of bookings, average ADR, and total stay nights.

### Code Explanation

- `groupby("hotel")` groups the records according to hotel type.
- `Booking_Count` calculates the total number of bookings.
- `Average_ADR` calculates the average daily rate for each hotel type.
- `Total_Stay_Nights` calculates the total number of stay nights.
- `reset_index()` converts the grouped result back into a DataFrame.

These aggregated features provide additional domain-level information that can be considered during feature selection.

## 8. Create Interaction Features

Interaction features combine two or more existing features to capture relationships that may not be visible from individual features.

For the Hotel Bookings dataset, useful interaction features can be created between:

- Lead time and ADR
- Total stay nights and ADR
- Total guests and total stay nights

These interactions can help the model understand how different booking characteristics work together.

```python
df["LeadTime_ADR_Interaction"] = (
    df["lead_time"] * df["adr"]
)

df["Stay_ADR_Interaction"] = (
    df["Total_Stay_Nights"] * df["adr"]
)

df["Guests_Stay_Interaction"] = (
    df["Total_Guests"] * df["Total_Stay_Nights"]
)

df[
    [
        "lead_time",
        "adr",
        "Total_Stay_Nights",
        "Total_Guests",
        "LeadTime_ADR_Interaction",
        "Stay_ADR_Interaction",
        "Guests_Stay_Interaction"
    ]
].head()
```

### Observation

Interaction features have been created by combining important booking characteristics.

These features represent relationships between booking lead time, pricing, stay duration, and guest count.

### Code Explanation

- `LeadTime_ADR_Interaction` combines lead time and ADR.
- `Stay_ADR_Interaction` combines stay duration and ADR.
- `Guests_Stay_Interaction` combines guest count and stay duration.
- Multiplication is used to capture the combined effect of the features.
- These interaction features will be evaluated during the feature selection and importance stages.

## 9. Perform Feature Selection

Feature selection identifies the most useful features for the machine learning model and removes features that are irrelevant or redundant.

For the Hotel Bookings dataset, we will focus on numerical features that are available at booking time.

The target variable is `is_canceled`.

Feature selection helps to:

- Reduce unnecessary features
- Reduce model complexity
- Improve training efficiency
- Reduce the effect of irrelevant information
- Create a cleaner ML-ready feature set

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

selected_features = [
    "lead_time",
    "adults",
    "children",
    "babies",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests",
    "Total_Stay_Nights",
    "Total_Guests",
    "Estimated_Stay_Revenue",
    "Guests_Per_Night",
    "ADR_Per_Guest",
    "Special_Request_Rate",
    "LeadTime_ADR_Interaction",
    "Stay_ADR_Interaction",
    "Guests_Stay_Interaction"
]

X = df[selected_features].replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0)

y = df["is_canceled"]

selector = SelectKBest(
    score_func=mutual_info_classif,
    k=10
)

X_selected = selector.fit_transform(X, y)

selected_feature_names = X.columns[selector.get_support()]

selected_feature_names
```

### Observation

Feature selection was performed on the engineered numerical features.

The Mutual Information method was used to identify features that provide useful information about the target variable, `is_canceled`.

The top 10 selected features are retained for further analysis.

### Code Explanation

- `selected_features` contains the original and engineered numerical features.
- Infinite values are replaced with missing values and then filled with 0.
- `X` contains the input features and `y` contains the cancellation target.
- `SelectKBest` selects the best features based on the specified scoring method.
- `mutual_info_classif` measures the dependency between each feature and the classification target.
- `get_support()` identifies the features selected by the method.

These selected features will be further evaluated using feature importance and leakage analysis.

## 10. Analyze Feature Importance

Feature importance helps us understand which features contribute most to the machine learning model's predictions.

After feature selection, we can use a tree-based model to measure the importance of the selected features.

A Random Forest classifier is used because it can capture nonlinear relationships and provides feature importance values.

The target variable is `is_canceled`.

```python
from sklearn.ensemble import RandomForestClassifier

X_selected_df = X[selected_feature_names]
y = df["is_canceled"]

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_selected_df, y)

importance_df = pd.DataFrame({
    "Feature": X_selected_df.columns,
    "Importance": rf_model.feature_importances_
}).sort_values(
    by="Importance",
    ascending=False
)

importance_df
```

### Observation

The Random Forest model provides an importance value for each selected feature.

Features with higher importance values contributed more to the model's predictions within this fitted model.

### Code Explanation

- `RandomForestClassifier` creates a Random Forest classification model.
- The model is trained using the selected features and `is_canceled` as the target.
- `feature_importances_` returns the importance value of each feature.
- A DataFrame is created to organize the feature importance results.
- `sort_values()` arranges the features from highest to lowest importance.

Feature importance will be considered along with leakage and feature-selection results before creating the final ML-ready dataset.

## 11. Check for Feature Leakage

Feature leakage occurs when information that would not be available at prediction time is used as a model feature.

For the Hotel Bookings dataset, special attention is required for features related to the final reservation outcome.

Potential leakage features include:

- reservation_status
- reservation_status_date

These columns describe information after the booking outcome and should not be used when predicting whether a booking will be canceled.

The engineered features should also be checked to ensure they are created only from information available at prediction time.

```python
leakage_features = [
    "reservation_status",
    "reservation_status_date"
]

available_leakage_features = [
    col for col in leakage_features
    if col in df.columns
]

available_leakage_features
```

### Observation

The dataset contains `reservation_status` and `reservation_status_date`.

These features are related to the final reservation outcome and therefore should not be used when predicting `is_canceled`.

The engineered features created in this workflow are based on booking, guest, stay, pricing, and arrival information that can be available before the final outcome.

### Code Explanation

- `leakage_features` contains columns identified as potential post-outcome features.
- The list comprehension checks which of these columns are present in the dataset.
- These features will be removed before creating the final ML-ready feature dataset.

### Important

Feature leakage can make a machine learning model appear highly accurate during evaluation while performing poorly when used on genuinely new bookings.

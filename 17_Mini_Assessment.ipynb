# Section 1 – Conceptual Questions with Answers

### 1. What is Feature Engineering?

**Answer:**  
Feature Engineering is the process of creating new features or modifying existing features from raw data to make the data more useful for machine learning models. It can include transformations, aggregations, encoding, binning, and creating interaction features.

### 2. What is the difference between Feature Engineering and Feature Selection?

**Answer:**  
Feature Engineering focuses on creating or transforming features to improve the representation of the data. Feature Selection focuses on selecting the most useful features and removing irrelevant or redundant features.

**Example:** Creating `total_spend` from `price` and `quantity` is Feature Engineering. Selecting `total_spend` and removing an irrelevant column is Feature Selection.

### 3. Why is domain knowledge important in Feature Engineering?

**Answer:**  
Domain knowledge helps us understand which features are meaningful for a particular business problem. It helps in creating useful features, identifying irrelevant features, and detecting possible data leakage.

**Example:** In a hotel booking dataset, `total_stay_nights` can be useful because the length of stay may provide meaningful information about a booking.

### 4. What is Feature Leakage? Give an example.

**Answer:**  
Feature Leakage happens when information that would not be available at prediction time is used as an input feature. This can make the model perform very well during training or testing but poorly in real-world use.

**Example:** For predicting whether a hotel booking will be canceled, using `reservation_status` can cause leakage because it contains information about what happened after the booking outcome.

```python
df["Price_Quantity"] = df["price"] * df["quantity"]



### 6. What is Binning or Discretization?

**Answer:**
Binning is the process of converting a continuous numerical feature into a smaller number of meaningful groups or categories.

**Example:**
`Age = 25` → `Young Adult`

---

### 7. What is Feature Cardinality?

**Answer:**
Feature cardinality refers to the number of unique values present in a feature.

**Example:**
A `gender` column with Male and Female has low cardinality, while a `customer_id` column with mostly unique values has high cardinality.

---

### 8. Explain Target Encoding.

**Answer:**
Target Encoding replaces each category with a value calculated from the target variable, commonly the mean target value for that category.

It should be performed carefully because incorrect target encoding can cause data leakage.

---

### 9. Explain Frequency Encoding.

**Answer:**
Frequency Encoding replaces each categorical value with the number of times it appears in the dataset.

**Example:**
If "Electronics" appears 500 times, its frequency-encoded value will be 500.

---

### 10. When should Log Transformation be used?

**Answer:**
Log Transformation is commonly used when a numerical feature is highly right-skewed. It helps reduce the effect of very large values and can make the distribution more suitable for some machine learning algorithms.

**Example:**
Income, sales, price, and transaction amounts can be log-transformed when they have strong right skew.

---

### 11. What is the Curse of Dimensionality?

**Answer:**
The Curse of Dimensionality refers to the problems that can occur when the number of features becomes very large.

As the number of features increases:

* Data becomes more sparse.
* Models may require more data.
* Computation can increase.
* Irrelevant features can make learning more difficult.

---

### 12. Explain PCA.

**Answer:**
PCA, or Principal Component Analysis, is a dimensionality reduction technique that transforms many correlated features into a smaller number of new features called principal components.

The principal components capture as much important variation from the original data as possible.

---

### 13. What is Feature Importance?

**Answer:**
Feature Importance measures how useful or influential each feature is for a particular machine learning model.

For example, tree-based models such as Random Forest can provide feature importance values.

---

### 14. Why is Feature Selection important before model training?

**Answer:**
Feature Selection helps keep the most useful features and remove irrelevant or redundant features.

It can:

* Reduce model complexity.
* Improve computational efficiency.
* Reduce noise.
* Help reduce overfitting.
* Make the model easier to understand.

---

### 15. What is the difference between Filter, Wrapper, and Embedded Feature Selection methods?

**Answer:**

**Filter Methods:**
Select features using statistical properties of the data without depending on a specific machine learning model.

**Examples:** Correlation, Chi-square, Mutual Information, ANOVA.

**Wrapper Methods:**
Select features by testing different combinations or subsets of features using a machine learning model.

**Examples:** Recursive Feature Elimination (RFE), Sequential Feature Selection.

**Embedded Methods:**
Perform feature selection as part of the model training process.

**Examples:** Lasso Regression, Decision Tree, Random Forest.

**In simple terms:**

* **Filter →** Statistical methods select the features.
* **Wrapper →** Model performance is used to select features.
* **Embedded →** Feature selection happens during model training.

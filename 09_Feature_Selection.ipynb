## 5. Redundant Features

Redundant Features are features that provide similar or overlapping information.

When two or more features contain highly similar information, keeping all of them may add unnecessary complexity to the model.

### Example

In the Hotel Booking dataset:

- `stays_in_weekend_nights`
- `stays_in_week_nights`

are related because both describe the length of a customer's stay.

We can also create:

`total_stay_nights = stays_in_weekend_nights + stays_in_week_nights`

After creating the total stay feature, the original features may contain overlapping information depending on the model and objective.

### Why Remove Redundant Features?

Removing redundant features can:

- Reduce unnecessary information
- Reduce model complexity
- Improve model interpretability
- Reduce multicollinearity for suitable models
- Make the feature set more compact

Redundant features should be identified carefully because two related features may still contain different useful information.

```python
redundant_features = [
    "stays_in_weekend_nights",
    "stays_in_week_nights"
]

correlation_matrix = df[redundant_features].corr()

correlation_matrix
```

### Observation

The correlation matrix shows the relationship between `stays_in_weekend_nights` and `stays_in_week_nights`.

A high correlation between features can indicate that they contain overlapping information and may need further evaluation for redundancy.

### Code Explanation

- `redundant_features` contains the features being checked for similarity.
- `.corr()` calculates the correlation between the selected numerical features.
- The resulting matrix helps identify strongly related features.
- Correlation is only one way to identify redundancy; additional feature selection methods can be used for a more reliable decision.

## 6. Filter Methods

Filter Methods select features based on their statistical properties and relationship with the target variable.

They evaluate features independently of the machine learning model.

Common Filter Methods include:

- Variance Threshold
- Correlation-Based Selection
- Mutual Information
- Chi-Square

### Why Filter Methods are Useful

Filter methods can:

- Quickly evaluate many features
- Reduce the number of features before model training
- Be computationally efficient
- Work independently of a specific machine learning algorithm

### Advantages

- Fast and simple
- Model-independent
- Suitable for high-dimensional datasets
- Easy to understand and implement

### Limitations

- Usually evaluate features individually
- May miss interactions between features
- Selected features may not always produce the best model performance
- Some methods require specific data types or preprocessing

```python
from sklearn.feature_selection import VarianceThreshold

X = df[[
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]].fillna(0)

selector = VarianceThreshold(threshold=0)

X_filtered = selector.fit_transform(X)

selected_features = X.columns[selector.get_support()]

selected_features
```

### Observation

The Filter Method evaluates features using statistical characteristics without training a machine learning model.

In this example, Variance Threshold removes features with zero variance, meaning features that have the same value for every observation.

### Code Explanation

- `X` contains the numerical input features.
- `fillna(0)` handles missing values before feature selection.
- `VarianceThreshold(threshold=0)` removes features with zero variance.
- `fit_transform()` applies the feature selection.
- `get_support()` identifies the features that were retained.

## 7. Wrapper Methods

Wrapper Methods select features by repeatedly training a machine learning model and evaluating different combinations of features.

Instead of evaluating each feature independently, wrapper methods consider how a group of features works together.

A common example is Recursive Feature Elimination (RFE).

### How Wrapper Methods Work

Start with all features
↓
Train a model
↓
Evaluate feature importance
↓
Remove less important features
↓
Train the model again
↓
Repeat until the required number of features remains

### Advantages

- Considers interactions between features
- Uses actual model performance
- Can produce a strong feature subset for a specific model

### Limitations

- Computationally expensive
- Can take a long time with many features
- Depends on the selected machine learning model
- May be sensitive to the training dataset

```python
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier

X = df[[
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]].fillna(0)

y = df["is_canceled"]

model = RandomForestClassifier(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

rfe = RFE(
    estimator=model,
    n_features_to_select=4
)

rfe.fit(X, y)

selected_features = X.columns[rfe.support_]

selected_features
```

### Observation

The Wrapper Method evaluates different combinations of features using a machine learning model.

In this example, RFE repeatedly evaluates the features and selects the four features considered most useful by the Random Forest model.

### Code Explanation

- `RandomForestClassifier` is used as the underlying machine learning model.
- `RFE` performs Recursive Feature Elimination.
- `n_features_to_select=4` specifies that four features should remain.
- `fit()` trains the RFE process using the input features and target.
- `support_` identifies the features selected by RFE.

## 8. Embedded Methods

Embedded Methods perform feature selection as part of the machine learning model training process.

The model automatically identifies important features while learning from the data.

Common Embedded Methods include:

- L1/Lasso-Based Selection
- Tree-Based Feature Importance

### How Embedded Methods Work

Features
↓
Model Training
↓
Feature Importance / Coefficients
↓
Less Important Features Identified
↓
Selected Features

### Advantages

- Feature selection happens during model training
- Usually faster than Wrapper Methods
- Considers the relationship between features and the target
- Can work well with large datasets

### Limitations

- Depends on the selected model
- Different models may select different features
- Requires understanding of the model's feature importance or coefficients
- Some methods require appropriate preprocessing

```python
from sklearn.ensemble import RandomForestClassifier

X = df[[
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]].fillna(0)

y = df["is_canceled"]

model = RandomForestClassifier(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

model.fit(X, y)

feature_importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

feature_importance
```

### Observation

The Random Forest model assigns an importance score to each feature.

Features with higher importance scores contribute more to the model's predictions and can be considered stronger candidates for feature selection.

### Code Explanation

- `RandomForestClassifier` is used as the machine learning model.
- `fit()` trains the model using the selected features and target.
- `feature_importances_` provides the importance score of each feature.
- `pd.Series()` associates each importance score with its feature name.
- `sort_values()` arranges the features from highest to lowest importance.

## 9. Variance Threshold

Variance Threshold is a Filter Method used to remove features that have very little variation.

A feature with very low variance contains almost the same value for most observations and may provide limited useful information to the model.

### Example

If a feature has the same value for every booking, its variance is 0.

Such a feature does not help the model distinguish between different observations.

### Why Variance Threshold is Useful

It can help:

- Remove constant features
- Reduce the number of features
- Reduce dataset complexity
- Improve computational efficiency

### Advantages

- Simple and fast
- Does not require the target variable
- Useful as an initial feature-selection step
- Works well for removing constant features

### Limitations

- Does not consider the relationship with the target
- A low-variance feature can sometimes be useful
- The threshold must be selected carefully
- It does not detect redundant features by itself

```python
from sklearn.feature_selection import VarianceThreshold

X = df[[
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]].fillna(0)

selector = VarianceThreshold(threshold=0)

selector.fit(X)

selected_features = X.columns[selector.get_support()]

selected_features
```

### Observation

The Variance Threshold method checks the amount of variation in each feature.

With a threshold of `0`, features having zero variance are removed, while features with some variation are retained.

### Code Explanation

- `VarianceThreshold(threshold=0)` creates the feature selector.
- `fit()` calculates the variance of each feature.
- `get_support()` identifies the features that pass the threshold.
- `selected_features` contains the retained feature names.
- `fillna(0)` handles missing values before applying the method.

## 11. Mutual Information

Mutual Information is a Filter Method that measures how much information a feature provides about the target variable.

It can identify both linear and non-linear relationships between a feature and the target.

A higher Mutual Information score generally indicates that the feature contains more useful information about the target.

### Example

For the Hotel Booking dataset, Mutual Information can be used to measure the relationship between numerical booking features and `is_canceled`.

### Why Mutual Information is Useful

It can help:

- Identify informative features
- Detect non-linear relationships
- Rank features based on their information about the target
- Select the most useful features

### Advantages

- Can detect non-linear relationships
- Does not require a linear relationship
- Model-independent
- Useful for ranking features

### Limitations

- Scores are not always easy to interpret
- Can be computationally more expensive than simple correlation
- Results can depend on the dataset and estimation method
- Does not directly show whether the relationship is positive or negative

```python
from sklearn.feature_selection import mutual_info_classif

feature_cols = [
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]

X = df[feature_cols].fillna(0)
y = df["is_canceled"]

mi_scores = mutual_info_classif(
    X,
    y,
    random_state=42
)

mi_results = pd.Series(
    mi_scores,
    index=feature_cols
).sort_values(ascending=False)

mi_results
```

### Observation

The Mutual Information scores show how much information each feature provides about the target variable `is_canceled`.

Features with higher scores contain more information about the target and can be considered stronger candidates for feature selection.

### Code Explanation

- `mutual_info_classif()` calculates Mutual Information scores for classification problems.
- `X` contains the numerical input features.
- `y` contains the target variable `is_canceled`.
- `fillna(0)` handles missing values.
- The calculated scores are associated with the feature names using `pd.Series()`.
- `sort_values()` ranks the features from highest to lowest Mutual Information score.

## 12. Chi-Square

Chi-Square is a Filter Method used to measure the relationship between categorical features and a categorical target variable.

It checks whether the distribution of a categorical feature is significantly associated with the target.

### Example

For the Hotel Booking dataset, categorical features such as:

- `hotel`
- `meal`
- `deposit_type`
- `customer_type`

can be evaluated against the categorical target `is_canceled`.

### Why Chi-Square is Useful

Chi-Square can help:

- Identify relevant categorical features
- Measure the relationship between categorical variables
- Rank categorical features for selection
- Remove less informative categorical features

### Advantages

- Simple and fast
- Model-independent
- Useful for categorical features
- Suitable for classification problems

### Limitations

- Designed for categorical features
- Features should be non-negative
- Does not measure the strength or direction of the relationship directly
- Results can be affected by feature frequency and sample size

```python
from sklearn.feature_selection import chi2

feature_cols = [
    "hotel",
    "meal",
    "deposit_type",
    "customer_type"
]

X = pd.get_dummies(
    df[feature_cols],
    drop_first=False
).astype(int)

y = df["is_canceled"]

chi_scores, p_values = chi2(X, y)

chi_results = pd.DataFrame({
    "Feature": X.columns,
    "Chi_Square_Score": chi_scores,
    "P_Value": p_values
}).sort_values(
    "Chi_Square_Score",
    ascending=False
)

chi_results.head(10)
```

### Observation

The Chi-Square scores show the strength of association between the encoded categorical features and `is_canceled`.

A higher Chi-Square score indicates a stronger statistical association with the target.

The p-value can also be considered when determining whether the observed association is statistically significant.

### Code Explanation

- `pd.get_dummies()` converts categorical features into numerical binary columns.
- `.astype(int)` converts the encoded values into integers.
- `chi2()` calculates the Chi-Square score and p-value for each feature.
- `Chi_Square_Score` represents the calculated Chi-Square statistic.
- `P_Value` represents the statistical significance of the association.
- `sort_values()` ranks features from highest to lowest Chi-Square score.

## 13. Recursive Feature Elimination (RFE)

Recursive Feature Elimination (RFE) is a Wrapper Method that selects important features by repeatedly training a machine learning model and removing the least important features.

### How RFE Works

Start with all features
↓
Train the model
↓
Rank feature importance
↓
Remove the least important feature
↓
Train the model again
↓
Repeat until the required number of features remains

### Example

For the Hotel Booking dataset, RFE can be used with a Random Forest classifier to select the most useful features for predicting `is_canceled`.

### Advantages

- Considers the performance of a machine learning model
- Can identify a useful subset of features
- Considers features together rather than only individually

### Limitations

- Computationally expensive
- Requires repeated model training
- Performance depends on the selected estimator
- Can be slower when the dataset contains many features

```python
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier

feature_cols = [
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]

X = df[feature_cols].fillna(0)
y = df["is_canceled"]

model = RandomForestClassifier(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

rfe = RFE(
    estimator=model,
    n_features_to_select=4
)

rfe.fit(X, y)

rfe_results = pd.DataFrame({
    "Feature": X.columns,
    "Selected": rfe.support_,
    "Ranking": rfe.ranking_
}).sort_values("Ranking")

rfe_results
```

### Observation

RFE ranks the features according to their usefulness to the Random Forest model.

Features with a ranking of `1` are the features selected by RFE. The remaining features receive higher ranking values based on the order in which they were eliminated.

### Code Explanation

- `RFE` performs Recursive Feature Elimination.
- `RandomForestClassifier` is used as the estimator.
- `n_features_to_select=4` tells RFE to retain four features.
- `fit()` trains the RFE process using the input features and target.
- `support_` indicates whether each feature was selected.
- `ranking_` provides the ranking assigned to each feature.
- `rfe_results` combines the selection status and ranking into a table.

## 14. L1/Lasso-Based Selection

L1 regularization, commonly known as Lasso, can be used as an Embedded Method for feature selection.

Lasso adds a penalty to the model coefficients. As a result, some feature coefficients can become exactly zero.

Features with zero coefficients can be removed because they do not contribute to the model under the selected regularization strength.

### Example

For the Hotel Booking dataset, L1-based Logistic Regression can be used to select features for predicting `is_canceled`.

### Why L1/Lasso-Based Selection is Useful

It can help:

- Automatically remove less useful features
- Reduce the number of features
- Produce a simpler model
- Perform feature selection during model training

### Advantages

- Performs feature selection during training
- Can produce sparse models
- Helps reduce model complexity
- Useful when there are many features

### Limitations

- Requires appropriate feature scaling
- Selection depends on the regularization strength
- With highly correlated features, L1 may select one feature and ignore another similar feature
- Primarily useful with models that support L1 regularization

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectFromModel

feature_cols = [
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]

X = df[feature_cols].fillna(0)
y = df["is_canceled"]

model = Pipeline([
    ("scaler", StandardScaler()),
    ("logistic", LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        random_state=42
    ))
])

model.fit(X, y)

coefficients = model.named_steps["logistic"].coef_[0]

l1_results = pd.DataFrame({
    "Feature": feature_cols,
    "Coefficient": coefficients,
    "Absolute_Coefficient": np.abs(coefficients)
}).sort_values(
    "Absolute_Coefficient",
    ascending=False
)

l1_results
```

### Observation

L1 regularization shrinks some model coefficients toward zero.

Features with coefficients close to zero contribute less to the model, while features with larger absolute coefficients have a stronger influence on the prediction.

### Code Explanation

- `StandardScaler()` standardizes the numerical features before applying L1 regularization.
- `LogisticRegression()` is used because it supports L1 regularization.
- `penalty="l1"` applies L1 regularization.
- `C=0.1` controls the strength of regularization.
- `fit()` trains the model.
- `coef_` returns the learned feature coefficients.
- `np.abs()` calculates the absolute coefficient values.
- `l1_results` ranks the features according to their absolute coefficient values.

## 15. Tree-Based Feature Importance

Tree-Based Feature Importance is an Embedded Method used by tree-based machine learning models to measure the importance of features.

Models such as Decision Trees and Random Forests can calculate how much each feature contributes to making predictions.

### Example

For the Hotel Booking dataset, a Random Forest model can be trained to predict `is_canceled`.

The model provides an importance score for each feature.

Higher importance scores indicate that the feature contributed more to the model's decisions.

### Why Tree-Based Feature Importance is Useful

It can help:

- Identify important features
- Rank features
- Remove less useful features
- Understand which features influence the model

### Advantages

- Built directly into many tree-based models
- Can capture non-linear relationships
- Can capture feature interactions
- Does not require feature scaling

### Limitations

- Importance can be biased toward certain features
- Correlated features can share or distort importance
- Importance scores do not explain the direction of the relationship
- Results depend on the selected tree-based model

```python
from sklearn.ensemble import RandomForestClassifier

feature_cols = [
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]

X = df[feature_cols].fillna(0)
y = df["is_canceled"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X, y)

tree_importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

tree_importance
```

### Observation

The Random Forest model assigns an importance score to each feature.

Features with higher importance scores contributed more to the model's decision-making process and can be considered stronger candidates for feature selection.

### Code Explanation

- `RandomForestClassifier` creates the tree-based model.
- `n_estimators=100` uses 100 decision trees.
- `fit()` trains the model using the selected features and target.
- `feature_importances_` returns the importance score of each feature.
- `pd.Series()` associates the scores with their feature names.
- `sort_values()` ranks the features from highest to lowest importance.

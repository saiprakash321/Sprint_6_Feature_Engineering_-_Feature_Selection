# Module 5 - Aggregation Features

## 1. Count

### Explanation

Count calculates how many booking records exist for a particular group.

In the Hotel Booking dataset, we can count the number of bookings for each hotel.

### Example

City Hotel → Number of bookings = total bookings received by City Hotel

### Why we use / need

Count helps measure booking activity and demand.

A higher booking count can indicate greater popularity or demand for a hotel.

### When we use

Use Count when measuring:

- Number of bookings
- Booking activity
- Hotel demand
- Number of reservations by category

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# Count bookings for each hotel

hotel_booking_count = (
    df.groupby("hotel")["is_canceled"]
    .count()
    .reset_index()
)

hotel_booking_count.columns = [
    "hotel",
    "BookingCount"
]

hotel_booking_count
```

### Observation

The number of booking records has been calculated for each hotel type.

This aggregation provides a simple measure of booking activity for City Hotel and Resort Hotel.

## 2. Sum

### Explanation

Sum adds the values of a numerical feature for each group.

In the Hotel Booking dataset, we can calculate the total number of nights booked for each hotel.

The total stay nights are calculated as:

Total Stay Nights = Weekend Nights + Week Nights

### Example

City Hotel → Sum of all stay nights across its bookings

### Why we use / need

Total stay nights provide a measure of the overall demand and occupancy contribution of each hotel.

This can help identify which hotel type receives bookings with a higher total number of stay nights.

### When we use

Use Sum for:

- Total stay nights
- Total guests
- Total special requests
- Total bookings-related numerical values

```python
# Create total stay nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Calculate total stay nights for each hotel

hotel_total_stays = (
    df.groupby("hotel")["total_stay_nights"]
    .sum()
    .reset_index()
)

hotel_total_stays.columns = [
    "hotel",
    "TotalStayNights"
]

hotel_total_stays
```

### Observation

The total number of stay nights has been calculated for each hotel type.

This aggregation summarizes the overall number of nights associated with bookings for each hotel and can provide a useful measure of booking demand.

City Hotel → Average ADR across all bookings

Resort Hotel → Average ADR across all bookings

## 4. Median

### Explanation

Median finds the middle value of a numerical feature for each group.

Unlike the mean, the median is less affected by extreme values or outliers.

In the Hotel Booking dataset, we can calculate the median daily rate (`adr`) for each hotel.

### Example

If ADR values are [50, 60, 500]:

→ Median = 60

The extreme value 500 has less effect on the median than on the mean.

### Why we use / need

Hotel pricing data can contain unusually high or low values.

The median provides a more typical representation of the daily rate when the data contains extreme values.

### When we use

Use Median when:

- Numerical data contains outliers.
- Data is highly skewed.
- A typical value is required without strong influence from extreme observations.

# Calculate median ADR for each hotel

hotel_median_adr = (
    df.groupby("hotel")["adr"]
    .median()
    .reset_index()
)

hotel_median_adr.columns = [
    "hotel",
    "MedianADR"
]

hotel_median_adr

### Observation

The median daily rate has been calculated for each hotel type.

The median provides a typical ADR value while being less affected by extreme pricing values.

This makes it useful when hotel pricing data is skewed or contains outliers.

## 7. Standard Deviation

### Explanation

Standard Deviation measures how much a numerical value varies within each group.

A low standard deviation means the values are relatively consistent, while a high standard deviation means the values vary more widely.

In the Hotel Booking dataset, we can calculate the standard deviation of `adr` for each hotel.

### Example

ADR values [100, 102, 98]:

→ Low Standard Deviation

ADR values [50, 200, 20]:

→ High Standard Deviation

### Why we use / need

Standard deviation helps identify how consistent or variable hotel pricing is.

This can provide additional information beyond the average ADR.

### When we use

Use Standard Deviation when analyzing:

- Pricing variability
- Stay duration variability
- Guest count variability
- Consistency of numerical behavior within groups

```python
# Calculate standard deviation of ADR for each hotel

hotel_std_adr = (
    df.groupby("hotel")["adr"]
    .std()
    .reset_index()
)

hotel_std_adr.columns = [
    "hotel",
    "StdADR"
]

hotel_std_adr
```

### Observation

The standard deviation of ADR has been calculated for each hotel type.

A higher standard deviation indicates greater variation in daily rates, while a lower value indicates more consistent pricing within the hotel group.

## 8. Group-Based Aggregation

### Explanation

Group-Based Aggregation calculates multiple statistical measures for different groups at the same time.

Instead of calculating Count, Sum, Mean, Minimum, Maximum, and Standard Deviation separately, the `.agg()` function can calculate them together.

In the Hotel Booking dataset, we can create multiple summary statistics for each hotel.

### Example

For each hotel:

- Count of bookings
- Total stay nights
- Average ADR
- Minimum ADR
- Maximum ADR
- Standard deviation of ADR

### Why we use / need

Group-Based Aggregation creates a complete summary of each group in one step.

This is useful when creating a feature table for Machine Learning.

### When we use

Use Group-Based Aggregation when several statistical features are required for the same group.

```python
# Create total stay nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Calculate multiple aggregations for each hotel

hotel_agg = (
    df.groupby("hotel")
    .agg(
        BookingCount=("is_canceled", "count"),
        TotalStayNights=("total_stay_nights", "sum"),
        AverageADR=("adr", "mean"),
        MinimumADR=("adr", "min"),
        MaximumADR=("adr", "max"),
        StdADR=("adr", "std")
    )
    .reset_index()
)

hotel_agg
```

### Observation

Multiple aggregation features have been calculated for each hotel in a single operation.

The resulting table provides information about booking volume, total stay nights, average pricing, pricing range, and pricing variability.

This creates a compact summary that can be useful for Machine Learning feature development.

## 9. Customer-Level Aggregation

### Explanation

Customer-Level Aggregation summarizes multiple transactions or bookings to create features that describe the overall behavior of each customer.

The Hotel Booking dataset does not contain a `CustomerID` column. Therefore, true customer-level aggregation cannot be performed reliably using this dataset.

Instead, we can use `country` as a grouping variable to demonstrate the same aggregation concept.

### Example

For each country:

- Number of bookings
- Total stay nights
- Average ADR
- Number of unique room types

### Why we use / need

Group-level features summarize detailed booking records into meaningful behavioral information.

In a dataset containing customer identifiers, these features can be used for customer segmentation, retention analysis, and customer-level prediction.

### When we use

Use Customer-Level Aggregation when the dataset contains:

- A customer identifier
- Multiple transactions or bookings per customer
- Customer behavior that needs to be summarized

```python
# Customer-level aggregation is not possible because
# the Hotel Booking dataset has no CustomerID.

# Demonstrate the same aggregation concept using country

country_features = (
    df.groupby("country")
    .agg(
        BookingCount=("is_canceled", "count"),
        TotalStayNights=("total_stay_nights", "sum"),
        AverageADR=("adr", "mean"),
        UniqueRoomTypes=("reserved_room_type", "nunique")
    )
    .reset_index()
)

country_features.head()
```

### Observation

The Hotel Booking dataset does not contain a customer identifier, so customer-level aggregation cannot be created directly.

The same aggregation concept has been demonstrated using `country` as a grouping variable.

In a customer transaction dataset, the grouping column would be `CustomerID`, allowing customer behavior to be summarized into Machine Learning features.

## 10. Product-Level Aggregation

### Explanation

Product-Level Aggregation summarizes information for each product.

The Hotel Booking dataset does not contain a product identifier such as `ProductID` or `StockCode`.

Therefore, true product-level aggregation cannot be performed using this dataset.

Instead, we can demonstrate the same aggregation concept using `reserved_room_type` as a hotel booking category.

### Example

For each reserved room type:

- Number of bookings
- Total stay nights
- Average ADR
- Maximum ADR

### Why we use / need

Product-level aggregation helps summarize the performance and behavior associated with each product.

In a hotel dataset, room-type-level aggregation can provide similar information about booking demand and pricing.

### When we use

Use Product-Level Aggregation when the dataset contains product information and the ML task involves:

- Product performance
- Demand forecasting
- Recommendation systems
- Inventory planning

```python
# Demonstrate product-level aggregation
# using reserved room type

room_features = (
    df.groupby("reserved_room_type")
    .agg(
        BookingCount=("is_canceled", "count"),
        TotalStayNights=("total_stay_nights", "sum"),
        AverageADR=("adr", "mean"),
        MaximumADR=("adr", "max")
    )
    .reset_index()
)

room_features.head()
```

### Observation

The `reserved_room_type` feature has been used to demonstrate product-level aggregation because the Hotel Booking dataset does not contain a product identifier.

The resulting features summarize booking volume, total stay nights, and pricing patterns for each reserved room type.

This provides useful information about room-type performance and demand.

## 11. Transaction-Level Aggregation

### Explanation

Transaction-Level Aggregation summarizes information for each individual booking or reservation instead of summarizing customers or room types.

In the Hotel Booking dataset, each row represents a booking record. We can create a booking-level summary using the booking's stay and guest information.

### Example

For each booking:

- Total Stay Nights
- Total Guests
- Total Special Requests
- ADR

### Why we use / need

Transaction-level features help the model understand the characteristics of each individual booking.

They can be useful for analyzing booking value, stay size, guest composition, and booking behavior.

### When we use

Use Transaction-Level Aggregation when the ML task is related to:

- Booking-level prediction
- Cancellation prediction
- Booking value analysis
- Stay behavior analysis

```python
# Create total stay nights
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Create total guests
df["total_guests"] = (
    df["adults"] +
    df["children"].fillna(0) +
    df["babies"]
)

# Create transaction-level features
transaction_features = df[
    [
        "hotel",
        "total_stay_nights",
        "total_guests",
        "total_of_special_requests",
        "adr"
    ]
].copy()

transaction_features.head()
```

### Observation

Transaction-level features have been created for each hotel booking.

The resulting features summarize the stay duration, total number of guests, special requests, and daily rate for each booking.

These features provide useful information when building Machine Learning models for booking-level prediction.

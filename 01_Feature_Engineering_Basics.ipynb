## 1. What is a Feature?

A **feature** is an individual input variable or attribute used by a Machine Learning model to understand the data and make predictions.

In the Hotel Booking Demand dataset, examples of features include:

- `lead_time` – Number of days between booking and arrival.
- `adults` – Number of adults in the booking.
- `children` – Number of children.
- `adr` – Average Daily Rate of the booking.
- `market_segment` – Market segment through which the booking was made.
- `deposit_type` – Type of deposit made for the booking.
- `total_of_special_requests` – Number of special requests made by the customer.

These features provide information that can help a Machine Learning model identify patterns related to hotel booking cancellations.

```python
feature_columns = [
    "lead_time",
    "adults",
    "children",
    "adr",
    "market_segment",
    "deposit_type",
    "total_of_special_requests"
]

df[feature_columns].head()
```

### Observation

The selected columns represent different characteristics of hotel bookings.

Numerical features contain measurable values, while categorical features represent different groups or categories. These features can be used as inputs for Machine Learning models.

## 2. Feature vs Variable

A **variable** is any column or attribute that stores information about an observation in a dataset.

A **feature** is a variable that is selected or prepared to be used as an input to a Machine Learning model.

Therefore, every feature is a variable, but not every variable is necessarily used as a feature.

For example, `arrival_date_month` is a variable in the Hotel Booking Demand dataset. It can be converted into a numerical representation such as the arrival month number, which can be used as a Machine Learning feature.

```python
month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["arrival_month_num"] = df["arrival_date_month"].map(month_mapping)

df[["arrival_date_month", "arrival_month_num"]].head()
```

### Observation

`arrival_date_month` is the original variable, while `arrival_month_num` is a transformed feature created from it.

The numerical representation can be more convenient for certain Machine Learning algorithms and further feature engineering operations.

## 3. Feature vs Target

A **feature** is an input variable used by a Machine Learning model to make a prediction.

A **target** is the output variable that the model is trained to predict.

In the Hotel Booking Demand dataset, `is_canceled` is the target variable.

- `is_canceled = 0` → Booking was not canceled.
- `is_canceled = 1` → Booking was canceled.

Examples of features that can be used to predict cancellation include `lead_time`, `adr`, `deposit_type`, `market_segment`, and `total_of_special_requests`.

```python
X = df[
    [
        "lead_time",
        "adults",
        "children",
        "adr",
        "total_of_special_requests"
    ]
]

y = df["is_canceled"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

### Observation

The feature variables are used as inputs to the Machine Learning model, while `is_canceled` is the target that the model attempts to predict.

Keeping features and the target clearly separated is important when preparing data for Machine Learning.

## 4. What is Feature Engineering?

**Feature Engineering** is the process of creating, transforming, selecting, or extracting useful features from existing data to improve the performance of Machine Learning models.

Raw hotel booking data may contain individual columns that do not directly represent useful business information.

For example, `stays_in_weekend_nights` and `stays_in_week_nights` are separate variables. They can be combined to create a new feature called `total_stay_nights`.

Feature Engineering helps convert raw data into meaningful information that Machine Learning models can use more effectively.

```python
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights"
    ]
].head()
```

### Observation

The new `total_stay_nights` feature combines two existing variables into a single meaningful feature.

This feature gives a clearer representation of the total duration of the customer's stay and may provide useful information for predicting booking behavior.

## 5. Why Feature Engineering?

Feature Engineering is important because raw data is not always in the most useful form for Machine Learning models.

Feature Engineering can help to:

- Create meaningful information from existing variables.
- Represent business relationships more clearly.
- Transform features into suitable formats.
- Reduce unnecessary or redundant information.
- Improve the ability of Machine Learning models to identify patterns.
- Incorporate domain knowledge into the dataset.

For the Hotel Booking Demand dataset, combining weekend and weekday stays into `total_stay_nights` provides a more meaningful representation of the customer's total stay duration.

```python
df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights",
        "is_canceled"
    ]
].head()
```

### Observation

The engineered feature `total_stay_nights` summarizes two related variables into one useful feature.

Such features can help Machine Learning models work with information that is closer to the actual business meaning of the data.

## 6. Importance of Domain Knowledge

**Domain Knowledge** means understanding the specific business area or industry from which the data is collected.

Domain knowledge is important in Feature Engineering because it helps us understand what the columns represent, which relationships are meaningful, and which new features can provide useful information to a Machine Learning model.

In the Hotel Booking Demand dataset, understanding hotel booking behavior helps us identify meaningful features such as:

- `lead_time` – A longer booking lead time may be associated with different cancellation behavior.
- `deposit_type` – The type of deposit can provide information about booking commitment.
- `total_of_special_requests` – The number of special requests can represent customer engagement with the booking.
- `total_stay_nights` – The total duration of the stay provides a better representation of the booking than considering weekend and weekday nights separately.

Domain knowledge also helps identify features that should not be used because they contain information that becomes available only after the booking outcome.

```python
# Create a domain-based feature

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights",
        "is_canceled"
    ]
].head()
```

### Observation

Domain knowledge helps us create features that have a meaningful business interpretation.

In this example, combining weekend and weekday nights gives the `total_stay_nights` feature, which represents the complete duration of a hotel stay.

## 7. Good Features vs Bad Features

Not every variable in a dataset is a good feature for Machine Learning.

A **good feature** provides useful information that is available at the time of prediction and has a meaningful relationship with the target.

A **bad feature** may contain irrelevant information, excessive missing values, redundant information, identifiers, or information that is only available after the target outcome.

For the Hotel Booking Demand dataset:

| Feature | Type | Reason |
|---|---|---|
| `lead_time` | Good | Provides information about booking timing. |
| `adr` | Good | Represents the average daily rate of the booking. |
| `deposit_type` | Good | Provides information about booking commitment. |
| `total_stay_nights` | Good | Represents the complete stay duration. |
| `is_canceled` | Bad as a feature | It is the target variable. |
| `reservation_status` | Bad | It contains information about the final booking outcome. |
| `reservation_status_date` | Bad | It is related to the final reservation status and can cause leakage. |

A feature should be evaluated based on its usefulness, availability at prediction time, and risk of data leakage.

```python
# Compare examples of potentially useful and problematic columns

good_features = [
    "lead_time",
    "adr",
    "deposit_type",
    "total_stay_nights"
]

potentially_bad_features = [
    "is_canceled",
    "reservation_status",
    "reservation_status_date"
]

print("Good Feature Examples:")
print(df[good_features].head())

print("\nPotentially Bad Feature Examples:")
print(df[potentially_bad_features].head())
```

### Observation

Good features provide useful information that can be known when making the prediction.

`reservation_status` and `reservation_status_date` should not be used for predicting cancellation at the time of booking because they contain information related to the final reservation outcome.

The target variable `is_canceled` must also remain separate from the input features.

## 8. Feature Engineering Workflow

Feature Engineering follows a systematic process to convert raw data into useful Machine Learning features.

The typical workflow is:

**Raw Data → Understand Features → Create Features → Transform Features → Select Features → Check Leakage → ML-Ready Features**

For the Hotel Booking Demand dataset, the workflow can be applied as follows:

1. Understand the available hotel booking variables.
2. Identify the target variable `is_canceled`.
3. Create meaningful features from existing variables.
4. Transform features into suitable formats.
5. Select relevant features for the Machine Learning model.
6. Check for data leakage.
7. Prepare the final set of ML-ready features.

For example, `stays_in_weekend_nights` and `stays_in_week_nights` can be combined to create `total_stay_nights`.

```python
# Example of the Feature Engineering workflow

# Feature Creation
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Select relevant features
selected_features = [
    "lead_time",
    "adults",
    "children",
    "adr",
    "total_stay_nights",
    "total_of_special_requests"
]

feature_data = df[selected_features]

feature_data.head()
```

### Observation

The Feature Engineering workflow converts raw hotel booking variables into a more useful feature set.

In this example, `total_stay_nights` is created from two existing variables, and relevant features are selected for further Machine Learning preparation.

Feature creation, transformation, selection, and leakage checking are important steps before using the data for model training.

## 9. Feature Creation

**Feature Creation** is the process of creating new features from existing variables to represent useful information in a more meaningful way.

Feature creation can involve:

- Addition of two or more features.
- Subtraction between features.
- Multiplication or division.
- Ratios and percentages.
- Aggregation of related variables.
- Combining existing information based on domain knowledge.

In the Hotel Booking Demand dataset, `stays_in_weekend_nights` and `stays_in_week_nights` represent different parts of the customer's stay.

We can combine them to create `total_stay_nights`, which represents the complete number of nights booked.

Another useful feature is `total_guests`, created by combining the number of adults, children, and babies.

```python
# Feature 1: Total Stay Nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Feature 2: Total Guests

df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights",
        "adults",
        "children",
        "babies",
        "total_guests"
    ]
].head()
```

### Observation

Two new features were created from existing hotel booking variables.

- `total_stay_nights` represents the complete duration of the customer's stay.
- `total_guests` represents the total number of people included in the booking.

These engineered features provide a more meaningful representation of the original booking information and can be useful for Machine Learning models.

## 10. Feature Transformation

**Feature Transformation** is the process of changing the representation or scale of existing features so that they are more suitable for Machine Learning models.

Common feature transformation techniques include:

- Log transformation
- Square root transformation
- Power transformation
- Scaling
- Normalization
- Standardization

In the Hotel Booking Demand dataset, `lead_time` represents the number of days between booking and arrival. This feature can have a skewed distribution because some bookings are made much earlier than others.

A log transformation can reduce the effect of very large values and make the distribution more suitable for analysis and Machine Learning.

```python
# Log transformation of lead_time

df["lead_time_log"] = np.log1p(df["lead_time"])

df[["lead_time", "lead_time_log"]].head()
```

### Observation

The `lead_time_log` feature is a transformed version of `lead_time`.

The log transformation reduces the impact of very large values and can help handle skewness in numerical features.

In a Machine Learning workflow, transformations should be learned from the training data and then applied to validation and test data to avoid data leakage.

## 11. Feature Selection

**Feature Selection** is the process of selecting the most relevant features for a Machine Learning model and removing features that are unnecessary, redundant, or irrelevant.

Feature selection can help to:

- Reduce the number of input features.
- Remove irrelevant information.
- Reduce model complexity.
- Improve model performance.
- Make the model easier to interpret.
- Reduce the risk of overfitting.

Common Feature Selection methods include:

- **Filter Methods** – Select features using statistical measures such as correlation.
- **Wrapper Methods** – Evaluate different feature subsets using a Machine Learning model.
- **Embedded Methods** – Select important features during model training, such as feature importance from tree-based models.

For the Hotel Booking Demand dataset, features such as `lead_time`, `adr`, and `total_of_special_requests` may provide useful information for predicting `is_canceled`.

```python
# Select numerical features and check their correlation with the target

numeric_features = [
    "lead_time",
    "adults",
    "children",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr",
    "total_of_special_requests"
]

correlation = df[numeric_features + ["is_canceled"]].corr()["is_canceled"]

correlation.sort_values(ascending=False)
```

### Observation

The correlation values provide an initial indication of the relationship between numerical features and the target variable `is_canceled`.

Features with stronger relationships may be considered for further analysis, while features with very weak relationships may require additional investigation.

Correlation alone should not be used as the final feature selection method. More detailed feature selection techniques will be covered in the later Feature Selection notebook.

## 12. Feature Extraction

**Feature Extraction** is the process of deriving useful information from existing data and representing it as a new feature that can be used by a Machine Learning model.

Feature extraction is commonly used when the original data contains information in a form that is not directly useful for Machine Learning.

In the Hotel Booking Demand dataset, the arrival date information is represented using separate columns such as `arrival_date_year`, `arrival_date_month`, and `arrival_date_day_of_month`.

These columns can be combined to create a complete arrival date and useful information such as the day of the week can then be extracted from it.

```python
# Create a complete arrival date

df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str)
)

# Extract the day of the week

df["arrival_day_of_week"] = df["arrival_date"].dt.day_name()

df[
    [
        "arrival_date_year",
        "arrival_date_month",
        "arrival_date_day_of_month",
        "arrival_date",
        "arrival_day_of_week"
    ]
].head()
```

### Observation

The separate arrival date components were combined to create `arrival_date`.

The day of the week was then extracted from the date and stored as `arrival_day_of_week`.

This creates a new feature from existing information and provides a more useful representation of the arrival date for further analysis and Machine Learning.

## 13. Feature Leakage

**Feature Leakage** occurs when information that would not be available at the time of prediction is used as an input feature for a Machine Learning model.

Feature leakage can cause a model to achieve unrealistically high performance during training and testing, but perform poorly when used on new real-world data.

In the Hotel Booking Demand dataset, `reservation_status` and `reservation_status_date` contain information about the final status of a reservation.

If the goal is to predict whether a booking will be canceled at the time of booking, these columns should not be used as input features because their information becomes available after the booking outcome.

Therefore, these columns should be removed before model training.

```python
# Features that can cause leakage for cancellation prediction

leakage_features = [
    "reservation_status",
    "reservation_status_date"
]

df[leakage_features].head()
```

```python
# Remove leakage-prone features

df_ml = df.drop(columns=leakage_features)

print("Remaining columns:", df_ml.shape[1])
print("Leakage features present:",
      [col for col in leakage_features if col in df_ml.columns])
```

### Observation

`reservation_status` and `reservation_status_date` were identified as potential leakage features because they contain information related to the final reservation outcome.

These features should be excluded when building a model that predicts cancellation at the time of booking.

Preventing feature leakage ensures that the Machine Learning model learns from information that would actually be available when making a real-world prediction.

## Role of Feature Engineering in the ML Pipeline

Feature Engineering plays an important role in preparing data for Machine Learning.

The overall Machine Learning workflow can be represented as:

**Raw Data → Data Cleaning → Preprocessing → Feature Engineering → Feature Selection → Model Training → Model Evaluation → Prediction**

In the Hotel Booking Demand dataset:

- **Raw Data:** Original hotel booking information.
- **Data Cleaning:** Handle missing values, duplicates, and invalid values.
- **Preprocessing:** Encode categorical variables and prepare numerical features.
- **Feature Engineering:** Create meaningful features such as `total_stay_nights` and `total_guests`.
- **Feature Selection:** Select relevant features and remove unnecessary or leakage-prone features.
- **Model Training:** Train a Machine Learning model using the prepared features.
- **Model Evaluation:** Evaluate the model using appropriate performance metrics.
- **Prediction:** Use the trained model to predict whether a hotel booking will be canceled.

```python
# Example of the final feature set for the ML pipeline

final_features = [
    "lead_time",
    "adults",
    "children",
    "babies",
    "adr",
    "total_stay_nights",
    "total_guests",
    "total_of_special_requests"
]

X = df[final_features]
y = df["is_canceled"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

X.head()
```

### Observation

Feature Engineering connects raw business data with Machine Learning by converting existing information into meaningful and useful features.

In this notebook, features such as `total_stay_nights`, `total_guests`, `lead_time_log`, and `arrival_day_of_week` were created or transformed from the original hotel booking data.

The final feature set should contain only information that is appropriate and available at the time of prediction. Leakage-prone features such as `reservation_status` and `reservation_status_date` should be excluded.

Feature Engineering therefore helps prepare the dataset for Feature Selection and subsequent Machine Learning tasks.

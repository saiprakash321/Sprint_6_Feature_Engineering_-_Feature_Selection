## 2. Model-Based Feature Importance

Model-Based Feature Importance is a technique where a machine learning model is used to determine how important each feature is for making predictions.

Different machine learning models provide different ways of measuring feature importance.

For example:

- Decision Trees use the reduction in impurity.
- Random Forests combine importance across multiple trees.
- Linear models can use the magnitude of their coefficients.

### Example

For the Hotel Booking dataset, a machine learning model can be trained to predict `is_canceled`.

The model can then provide an importance score for each input feature.

### Why Model-Based Feature Importance is Useful

It can help:

- Identify important features
- Rank features
- Understand model behavior
- Support feature selection
- Improve model interpretability

### Advantages

- Directly related to the trained model
- Easy to calculate for supported models
- Useful for ranking features
- Can identify non-linear relationships with tree-based models

### Limitations

- Results depend on the selected model
- Different models may produce different importance rankings
- Importance does not imply causation
- Correlated features can affect importance scores

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X, y)

model_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

model_importance
```

### Observation

The Random Forest model assigns an importance score to each feature.

Features with higher importance scores contributed more to the model's prediction process.

The importance scores can be used to rank the features and identify features that may be useful for further analysis.

### Code Explanation

- `RandomForestClassifier` creates the machine learning model.
- `n_estimators=100` uses 100 decision trees.
- `fit()` trains the model using `X` and `y`.
- `feature_importances_` returns the importance score of each feature.
- `pd.DataFrame()` creates a table containing feature names and importance scores.
- `sort_values()` ranks the features from highest to lowest importance.

## 3. Decision Tree Feature Importance

A Decision Tree can calculate feature importance based on how much each feature helps reduce impurity when making decisions.

When a feature is used to split the data and creates better-separated groups, it receives a higher importance score.

### Example

For the Hotel Booking dataset, a Decision Tree can be trained to predict `is_canceled`.

The model can then provide an importance score for each booking feature.

### Why Decision Tree Feature Importance is Useful

It can help:

- Identify important features
- Rank features
- Understand which features are used by the tree
- Capture non-linear relationships

### Advantages

- Easy to calculate
- Easy to interpret
- Can capture non-linear relationships
- Does not require feature scaling

### Limitations

- Can be affected by overfitting
- Importance can be biased toward some features
- Correlated features can affect importance scores
- Importance does not show whether a feature increases or decreases the target

```python
from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

decision_tree.fit(X, y)

dt_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": decision_tree.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

dt_importance
```

### Observation

The Decision Tree assigns an importance score to each feature based on how much the feature contributes to reducing impurity during the tree's splitting process.

Features with higher importance scores have a greater contribution to the Decision Tree's predictions.

### Code Explanation

- `DecisionTreeClassifier` creates the Decision Tree model.
- `max_depth=5` limits the depth of the tree and helps control model complexity.
- `fit()` trains the Decision Tree using the input features and target.
- `feature_importances_` returns the importance score of each feature.
- `pd.DataFrame()` creates a table containing feature names and importance scores.
- `sort_values()` ranks the features from highest to lowest importance.

## 4. Random Forest Feature Importance

Random Forest Feature Importance measures how much each feature contributes to the predictions of a Random Forest model.

A Random Forest contains multiple Decision Trees. The importance of a feature is calculated by combining its contribution across the trees.

### Example

For the Hotel Booking dataset, a Random Forest can be trained to predict `is_canceled`.

The model can then rank features according to their importance.

### Why Random Forest Feature Importance is Useful

It can help:

- Identify important booking features
- Rank features
- Understand model behavior
- Support feature selection
- Capture non-linear relationships

### Advantages

- More stable than a single Decision Tree
- Can capture non-linear relationships
- Can handle feature interactions
- Does not require feature scaling

### Limitations

- Importance can be affected by correlated features
- Can favor features with many possible split points
- Does not show the direction of the relationship
- Importance does not imply causation

```python
from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest.fit(X, y)

rf_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": random_forest.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

rf_importance
```

### Observation

The Random Forest assigns an importance score to each feature based on its contribution across multiple decision trees.

Features with higher importance scores have a greater contribution to the Random Forest's predictions.

### Code Explanation

- `RandomForestClassifier` creates the Random Forest model.
- `n_estimators=100` uses 100 decision trees.
- `fit()` trains the model using the input features and target.
- `feature_importances_` returns the importance score for each feature.
- `pd.DataFrame()` creates a table containing the feature names and importance scores.
- `sort_values()` ranks the features from highest to lowest importance.

## 5. Permutation Importance

Permutation Importance measures feature importance by checking how much the model's performance decreases when the values of a feature are randomly shuffled.

If shuffling a feature causes a large decrease in model performance, that feature is considered important to the model.

### How Permutation Importance Works

Train the model
↓
Measure model performance
↓
Shuffle one feature
↓
Measure performance again
↓
Calculate the performance decrease
↓
Repeat for each feature

### Example

For the Hotel Booking dataset, permutation importance can be used with a Random Forest model to identify which features are most important for predicting `is_canceled`.

### Advantages

- Can be used with many different machine learning models
- Measures importance based on actual model performance
- Easy to compare features
- Can be more informative than built-in tree importance in some situations

### Limitations

- Can be computationally expensive
- Results can be affected by correlated features
- Requires a validation or test dataset for reliable evaluation
- Importance scores can vary between runs

```python
from sklearn.inspection import permutation_importance

perm_result = permutation_importance(
    random_forest,
    X,
    y,
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

permutation_results = pd.DataFrame({
    "Feature": X.columns,
    "Importance_Mean": perm_result.importances_mean,
    "Importance_Std": perm_result.importances_std
}).sort_values(
    "Importance_Mean",
    ascending=False
)

permutation_results
```

### Observation

The Permutation Importance results show how much the model's performance changes when each feature is randomly shuffled.

A higher mean importance indicates that shuffling the feature caused a larger decrease in model performance, suggesting that the feature is important to the model.

### Code Explanation

- `permutation_importance()` calculates the importance of each feature by randomly shuffling its values.
- `random_forest` is the trained Random Forest model.
- `n_repeats=5` repeats the permutation process five times for each feature.
- `importances_mean` gives the average importance score.
- `importances_std` shows the variation in the importance scores across repetitions.
- `sort_values()` ranks the features from highest to lowest mean importance.

## 6. Feature Importance vs Feature Selection

Feature Importance and Feature Selection are related but different concepts.

### Feature Importance

Feature Importance measures how much each feature contributes to a machine learning model's predictions.

It provides an importance score or ranking for the features.

### Feature Selection

Feature Selection is the process of choosing a subset of features and removing features that are irrelevant, redundant, or less useful.

### Example

For the Hotel Booking dataset:

Feature Importance:

`lead_time` → High importance  
`adr` → Medium importance  
`adults` → Lower importance

Feature Selection:

Keep the most useful features and remove features that do not provide enough predictive value.

### Key Difference

Feature Importance tells us:

"Which features are important to the model?"

Feature Selection tells us:

"Which features should we keep for the model?"

### Advantages of Using Feature Importance

- Helps understand model behavior
- Provides feature rankings
- Supports feature selection decisions

### Limitations

- Importance depends on the model or method used
- Different methods may produce different rankings
- High importance does not mean causation

```python
feature_comparison = pd.DataFrame({
    "Feature": X.columns,
    "Random_Forest_Importance": random_forest.feature_importances_,
    "Decision_Tree_Importance": decision_tree.feature_importances_
}).sort_values(
    "Random_Forest_Importance",
    ascending=False
)

feature_comparison
```

### Observation

Feature importance provides a ranking of features based on their contribution to different machine learning models.

The Random Forest and Decision Tree may produce different importance scores because they learn patterns differently.

These importance rankings can support feature selection, but the final feature set should be evaluated using model performance.

### Code Explanation

- `random_forest.feature_importances_` provides importance scores from the Random Forest.
- `decision_tree.feature_importances_` provides importance scores from the Decision Tree.
- `pd.DataFrame()` combines both sets of scores for comparison.
- `sort_values()` ranks the features according to Random Forest importance.

## 7. Interpreting Feature Importance

Interpreting Feature Importance means understanding the importance scores produced by a machine learning model.

A higher importance score generally means that the feature contributed more to the model's predictions.

### Example

Suppose the model produces the following results:

| Feature | Importance |
|---|---:|
| `lead_time` | 0.35 |
| `adr` | 0.20 |
| `adults` | 0.12 |
| `children` | 0.08 |

In this example, `lead_time` has the highest importance score among the listed features.

### Important Points

- Importance scores are relative to the model and dataset.
- A higher score means greater contribution to the model's predictions.
- A lower score does not always mean the feature is completely useless.
- Feature importance does not indicate whether a feature increases or decreases the target.
- Feature importance does not prove a causal relationship.

### Advantages

- Easy to understand feature rankings
- Helps identify potentially important variables
- Supports feature selection and model interpretation

### Limitations

- Scores depend on the model and method
- Correlated features can make interpretation difficult
- Importance does not explain the direction of an effect
- Importance does not imply causation

```python
importance_interpretation = (
    rf_importance
    .assign(
        Percentage=lambda x: x["Importance"] * 100
    )
)

importance_interpretation
```

### Observation

The feature importance scores show the relative contribution of each feature to the Random Forest model.

The `Percentage` column expresses the importance scores as percentages, making the relative contribution easier to compare.

### Code Explanation

- `rf_importance` contains the feature importance scores from the Random Forest model.
- `assign()` creates a new column called `Percentage`.
- The importance score is multiplied by 100 to convert it into a percentage.
- The resulting table helps compare the relative importance of the features.

## 8. Limitations of Feature Importance

Feature Importance is useful for understanding model behavior, but the importance scores should be interpreted carefully.

A high feature importance score does not necessarily mean that the feature has a direct or causal effect on the target variable.

### Common Limitations

- Importance scores depend on the machine learning model.
- Different models can produce different importance rankings.
- Correlated features can affect the importance scores.
- Some methods may favor features with more possible split points.
- Feature importance does not show whether a feature increases or decreases the target.
- Feature importance does not prove a causal relationship.
- A low importance score does not always mean that the feature is completely useless.
- Importance calculated on training data may not represent performance on unseen data.

### Hotel Booking Example

For predicting `is_canceled`, a feature may receive a high importance score because it helps the Random Forest make accurate predictions.

However, this does not mean that the feature directly causes a booking to be canceled.

Feature importance should therefore be combined with other methods such as correlation analysis, permutation importance, domain knowledge, and model evaluation.

### Observation

Feature importance is useful for ranking and understanding features, but it should not be treated as the only method for deciding which features to keep.

Different feature importance techniques can produce different results. Therefore, feature importance should be interpreted together with other feature selection methods and domain knowledge.

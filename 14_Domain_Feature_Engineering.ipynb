# 14. Domain-Based Feature Engineering

Domain chosen: Hotel Booking / Hospitality

Domain-based feature engineering means creating features using knowledge of how a particular business works.

In hotel booking data, useful features can be created from stay duration, guest information, booking details, pricing, and special requests.

For every feature, we will document:

- Feature Name
- Source Column(s)
- Logic
- Business Meaning
- ML Relevance
- Leakage Risk
- Retain or Remove

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df["children"] = df["children"].fillna(0)

df.head()
```

### Observation

The Hotel Bookings dataset has been loaded and prepared for domain-based feature engineering.

The dataset contains booking, guest, stay, pricing, and reservation-related information that can be converted into meaningful hospitality features.

### Code Explanation

- `pd.read_csv()` loads the Hotel Bookings dataset.
- Missing values in `children` are treated as 0.
- The dataset is now ready for creating domain-specific features.

### Domain Thinking

A useful hotel feature should represent a meaningful aspect of a booking or guest behavior and should only use information available at the time the prediction is made.

### Feature 1: Total_Stay_Nights

Feature Name: Total_Stay_Nights

Source Columns: stays_in_weekend_nights, stays_in_week_nights

Logic: Add the number of weekend nights and weekday nights.

Business Meaning: Represents the complete duration of a guest's stay.

ML Relevance: Useful for understanding booking duration and predicting cancellation, demand, or booking value.

Leakage Risk: Low, because the planned stay duration is available from the booking information.

Retain or Remove: Retain, because it is a meaningful hotel-domain feature.

```python
df["Total_Stay_Nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "Total_Stay_Nights"
    ]
].head()
```

### Observation

The two separate stay-duration columns have been combined into one meaningful feature, `Total_Stay_Nights`.

This feature gives a direct view of the total planned length of the hotel stay.

### Code Explanation

- `stays_in_weekend_nights` contains the number of weekend nights.
- `stays_in_week_nights` contains the number of weekday nights.
- Adding both columns creates `Total_Stay_Nights`.
- The feature is retained because it provides useful domain information without introducing post-outcome information.

### Feature 4: Is_Weekend_Stay

Feature Name: Is_Weekend_Stay

Source Columns: stays_in_weekend_nights

Logic: If the booking has one or more weekend nights, assign 1; otherwise assign 0.

Business Meaning: Indicates whether the guest's stay includes at least one weekend night.

### Code Explanation

- The feature is retained because weekend stay information can help represent hotel booking patterns.

### Feature 6: Booking_Lead_Category

Feature Name: Booking_Lead_Category

Source Columns: lead_time

Logic: Group the number of days between booking and arrival into meaningful categories:
- 0–30 days → Short Lead
- 31–90 days → Medium Lead
- More than 90 days → Long Lead

Business Meaning: Represents how early the guest made the booking before the arrival date.

ML Relevance: Booking lead time can capture different booking behaviors and may provide useful information for predicting cancellation and demand patterns.

Leakage Risk: Low, because lead time is available from the booking information.

Retain or Remove: Retain, because the categorical version makes lead-time patterns easier for an ML model to capture.

```python
df["Booking_Lead_Category"] = pd.cut(
    df["lead_time"],
    bins=[-1, 30, 90, np.inf],
    labels=["Short Lead", "Medium Lead", "Long Lead"]
)

df[
    [
        "lead_time",
        "Booking_Lead_Category"
    ]
].head()
```

### Observation

The continuous `lead_time` feature has been converted into three meaningful booking categories.

The categories are:
- Short Lead → 0–30 days
- Medium Lead → 31–90 days
- Long Lead → More than 90 days

### Code Explanation

- `pd.cut()` divides the numerical `lead_time` values into defined intervals.
- The `bins` define the boundaries for each category.
- `labels` assign meaningful names to the categories.
- The feature is retained because it represents different booking behaviors in an easy-to-understand form.

### Feature 7: Guests_Per_Night

Feature Name: Guests_Per_Night

Source Columns: Total_Guests, Total_Stay_Nights

Logic: Divide the total number of guests by the total number of stay nights. Zero stay nights are handled to avoid division by zero.

Business Meaning: Represents the average number of guests associated with each night of the booking.

ML Relevance: Helps describe booking size relative to stay duration and can provide additional information about booking patterns.

Leakage Risk: Low, because the feature uses booking-time information.

Retain or Remove: Retain, because it combines guest count and stay duration into a meaningful hospitality feature.

```python
df["Guests_Per_Night"] = np.where(
    df["Total_Stay_Nights"] > 0,
    df["Total_Guests"] / df["Total_Stay_Nights"],
    0
)

df[
    [
        "Total_Guests",
        "Total_Stay_Nights",
        "Guests_Per_Night"
    ]
].head()
```

### Observation

The `Guests_Per_Night` feature represents the number of guests relative to the total duration of the booking.

### Code Explanation

- `Total_Guests` provides the total number of guests.
- `Total_Stay_Nights` provides the total stay duration.
- The values are divided to create `Guests_Per_Night`.
- `np.where()` prevents division by zero when the stay duration is 0.
- The feature is retained because it combines two important hotel booking characteristics into one useful feature.

### Feature 8: ADR_Per_Guest

Feature Name: ADR_Per_Guest

Source Columns: adr, Total_Guests

Logic: Divide the Average Daily Rate (ADR) by the total number of guests. Zero guest counts are handled to avoid division by zero.

Business Meaning: Represents the average daily room rate relative to the number of guests in the booking.

ML Relevance: Helps capture the pricing level of a booking in relation to its guest count and can provide useful information for predicting booking behavior.

Leakage Risk: Low, provided ADR and guest count are available at the time of prediction.

Retain or Remove: Retain, because it provides a useful price-to-guest relationship.

```python
df["ADR_Per_Guest"] = np.where(
    df["Total_Guests"] > 0,
    df["adr"] / df["Total_Guests"],
    0
)

df[
    [
        "adr",
        "Total_Guests",
        "ADR_Per_Guest"
    ]
].head()
```

### Observation

The `ADR_Per_Guest` feature represents the Average Daily Rate relative to the number of guests in the booking.

### Code Explanation

- `adr` represents the Average Daily Rate.
- `Total_Guests` represents the total number of guests.
- Dividing ADR by the total guest count creates `ADR_Per_Guest`.
- `np.where()` prevents division by zero when there are no guests.
- The feature is retained because it adds a useful pricing perspective based on guest count.

### Feature 9: Estimated_Stay_Revenue

Feature Name: Estimated_Stay_Revenue

Source Columns: adr, Total_Stay_Nights

Logic: Multiply the Average Daily Rate (ADR) by the total number of stay nights.

Business Meaning: Provides an estimated room revenue for the complete stay based on the daily rate and planned stay duration.

ML Relevance: Helps represent the approximate monetary value of a booking and can provide useful information for predicting booking behavior.

Leakage Risk: Low, provided ADR and planned stay duration are available at the time of prediction.

Retain or Remove: Retain, because it combines pricing and stay duration into a meaningful business feature.

```python
df["Estimated_Stay_Revenue"] = (
    df["adr"] * df["Total_Stay_Nights"]
)

df[
    [
        "adr",
        "Total_Stay_Nights",
        "Estimated_Stay_Revenue"
    ]
].head()
```

### Observation

The `Estimated_Stay_Revenue` feature represents the approximate room revenue for the planned hotel stay.

### Code Explanation

- `adr` represents the Average Daily Rate.
- `Total_Stay_Nights` represents the total planned stay duration.
- Multiplying these values gives an estimated stay revenue.
- The feature is retained because it combines price and stay duration into one useful hospitality feature.
- This is an estimate based on ADR and does not necessarily represent the final amount charged to the guest.

### Feature 11: High_Demand_Period

Feature Name: High_Demand_Period

Source Columns: arrival_date_month

Logic: Group the arrival months into a simple seasonal indicator. Months commonly associated with higher booking activity are marked as 1, and the remaining months are marked as 0.

Business Meaning: Represents whether the booking falls within a potentially high-demand seasonal period.

ML Relevance: Seasonal booking patterns can help a model capture changes in hotel demand across different periods of the year.

Leakage Risk: Low, because the arrival month is known at the time of booking.

Retain or Remove: Retain, because seasonality is an important factor in the hospitality domain.

```python
high_demand_months = [
    "July",
    "August",
    "June",
    "September"
]

df["High_Demand_Period"] = (
    df["arrival_date_month"]
    .isin(high_demand_months)
    .astype(int)
)

df[
    [
        "arrival_date_month",
        "High_Demand_Period"
    ]
].head()
```

### Observation

The `High_Demand_Period` feature identifies bookings that fall within selected seasonal months.

The feature contains:
- 1 → Booking falls within the selected high-demand period
- 0 → Booking falls outside the selected period

### Code Explanation

- `high_demand_months` contains the selected months.
- `.isin()` checks whether each booking's arrival month belongs to this group.
- `.astype(int)` converts the result into 1 and 0.
- The feature is retained because arrival season can represent important demand patterns in hotel bookings.

Note: The selected months are a domain-based example, not a target-derived measure of actual demand.

### Feature 16: Special_Request_Rate

### Feature 17: Guest_Composition

Feature Name: Guest_Composition

Source Columns: adults, children, babies

Logic: Categorize the booking based on the combination of adults, children, and babies.

Business Meaning: Represents the type of guest group associated with the booking.

ML Relevance: Different guest compositions may have different booking patterns and requirements. This feature can help the model capture those differences.

Leakage Risk: Low, because guest composition is available in the booking information.

Retain or Remove: Retain, because guest composition is meaningful in the hotel booking domain.

```python
def get_guest_composition(row):
    if row["children"] > 0 or row["babies"] > 0:
        return "Family"
    elif row["adults"] == 1:
        return "Solo"
    else:
        return "Adults Only"

df["Guest_Composition"] = df.apply(
    get_guest_composition,
    axis=1
)

df[
    [
        "adults",
        "children",
        "babies",
        "Guest_Composition"
    ]
].head()
```

### Observation

The `Guest_Composition` feature categorizes bookings into meaningful guest groups:

- Family → Booking includes children or babies
- Solo → Booking has one adult and no children or babies
- Adults Only → Booking has multiple adults and no children or babies

### Code Explanation

- The function checks the number of adults, children, and babies.
- `apply()` applies the function to each booking.
- The resulting categories provide a higher-level representation of guest composition.
- The feature is retained because guest-group type is relevant to hotel booking behavior.

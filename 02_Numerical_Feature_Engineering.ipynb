# 1. Mathematical Transformations

Mathematical transformations combine two or more numerical columns using basic arithmetic operations to create a new feature that carries more meaning than either column alone.

**When to use:** whenever a raw column by itself does not capture the real quantity a model needs, but combining it with another column does.

**Why:** models may not automatically represent useful arithmetic relationships between columns, so explicitly creating these relationships can provide more meaningful features.

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# Example: Calculate total stay nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights"
    ]
].head()
```

### Observation

The `total_stay_nights` feature combines weekend and weekday nights into a single feature.

This provides a clearer representation of the customer's total stay duration than using the two individual columns separately.

# 2. Addition

Addition combines two or more numerical columns to represent a combined total.

**When to use:** when several columns represent different parts of a whole, such as combining weekend and weekday nights into the total length of a hotel stay.

**Why:** a single combined feature can represent the overall quantity more clearly and reduce the need for the model to infer the relationship between separate columns.

```python
# Create total stay nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights"
    ]
].head()
```

### Observation

The `total_stay_nights` feature combines `stays_in_weekend_nights` and `stays_in_week_nights` into one feature.

This feature gives a clearer representation of the total number of nights booked by a customer and can be useful for understanding booking behavior.

# 3. Subtraction

Subtraction between two numerical columns can be used to represent a gap or difference between two related values.

**When to use:** when the difference between two values provides more useful information than the individual values, such as measuring the difference between weekday and weekend stay nights.

**Why:** the difference directly represents the gap between two related quantities, which can help a Machine Learning model identify patterns more easily.

```python
# Calculate the difference between weekday and weekend stay nights

df["stay_nights_difference"] = (
    df["stays_in_week_nights"] -
    df["stays_in_weekend_nights"]
)

df[
    [
        "stays_in_week_nights",
        "stays_in_weekend_nights",
        "stay_nights_difference"
    ]
].head()
```

### Observation

The `stay_nights_difference` feature represents the difference between weekday and weekend nights.

A positive value indicates that the booking contains more weekday nights than weekend nights, while a negative value indicates more weekend nights.

# 4. Multiplication

Multiplication combines two numerical features to represent a combined effect or quantity.

**When to use:** when the actual business meaning depends on the product of two numerical variables.

**Why:** multiplying related features can create a meaningful feature that represents their combined effect and may provide additional information to a Machine Learning model.

```python
# Create total stay nights
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Create total guests
df["total_guests"] = (
    df["adults"] +
    df["children"].fillna(0) +
    df["babies"]
)

# Create total guest-nights
df["total_guest_nights"] = (
    df["total_guests"] * df["total_stay_nights"]
)

df[
    [
        "total_guests",
        "total_stay_nights",
        "total_guest_nights"
    ]
].head()
```

### Observation

The `total_guest_nights` feature combines the total number of guests and the total stay duration.

For example, a booking with 3 guests staying for 4 nights represents 12 guest-nights.

This feature provides a more meaningful measure of the overall size of a hotel booking.

# 5. Division

Division creates a new numerical feature by dividing one value by another to represent a value per unit.

**When to use:** when one numerical value needs to be compared relative to another value.

**Why:** division helps normalize data and creates meaningful features that are easier for Machine Learning models to understand.

In the Hotel Booking Demand dataset, we can calculate the **Average Daily Spend** by dividing the total booking amount by the total number of stay nights.

```python
# Calculate Average Daily Spend

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["booking_amount"] = df["adr"] * df["total_stay_nights"]

df["average_daily_spend"] = (
    df["booking_amount"] / df["total_stay_nights"].replace(0, 1)
)

df[
    [
        "adr",
        "total_stay_nights",
        "booking_amount",
        "average_daily_spend"
    ]
].head()
```

### Observation

The `average_daily_spend` feature represents the average amount spent per day during the hotel stay.

This feature is useful because it normalizes the booking amount based on the stay duration and provides a meaningful numerical feature for Machine Learning.

# 6. Ratios

A ratio expresses the relationship between two related numerical quantities, usually by dividing one quantity by another.

**When to use:** when the relative size of one quantity compared with another carries more useful information than their individual values.

**Why:** ratios make it easier to compare bookings with different sizes or durations and can provide scale-independent information to a Machine Learning model.

In the Hotel Booking Demand dataset, we can calculate the ratio of weekend stay nights to total stay nights.

```python
# Calculate weekend stay ratio

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["weekend_stay_ratio"] = (
    df["stays_in_weekend_nights"] /
    df["total_stay_nights"].replace(0, 1)
)

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights",
        "weekend_stay_ratio"
    ]
].head()
```

### Observation

The `weekend_stay_ratio` feature represents the proportion of the total stay that occurs during weekends.

A higher value indicates that a larger portion of the booking consists of weekend nights.

This ratio provides relative information about the booking duration and can be useful for identifying booking patterns.

# 7. Percentages

A percentage expresses a ratio as a value out of 100, making relative proportions easier to interpret.

**When to use:** when the proportion of a quantity is more meaningful than its absolute value, such as the percentage of a booking spent on weekend nights.

**Why:** percentages provide an intuitive and standardized scale that makes it easier to compare different bookings.

In the Hotel Booking Demand dataset, we can calculate the percentage of total stay nights that occur during weekends.

```python
# Calculate percentage of weekend stay nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["weekend_stay_percentage"] = (
    df["stays_in_weekend_nights"] /
    df["total_stay_nights"].replace(0, 1)
) * 100

df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "total_stay_nights",
        "weekend_stay_percentage"
    ]
].head()
```

### Observation

The `weekend_stay_percentage` feature represents the percentage of the customer's total stay that occurs during weekends.

This converts the weekend stay into an easy-to-interpret percentage and allows bookings with different stay durations to be compared more effectively.

# 8. Differences

A difference feature captures the change or gap between two related numerical values.

**When to use:** when the difference between two values provides more useful information than their individual values, such as the difference between weekday and weekend stay nights.

**Why:** the difference directly represents the magnitude and direction of change between two related quantities, which can help a Machine Learning model identify useful patterns.

In the Hotel Booking Demand dataset, we can calculate the difference between weekday and weekend stay nights.

```python
# Calculate the difference between weekday and weekend stay nights

df["stay_nights_difference"] = (
    df["stays_in_week_nights"] -
    df["stays_in_weekend_nights"]
)

df[
    [
        "stays_in_week_nights",
        "stays_in_weekend_nights",
        "stay_nights_difference"
    ]
].head()
```

### Observation

The `stay_nights_difference` feature represents the difference between weekday and weekend stay nights.

A positive value indicates more weekday nights, while a negative value indicates more weekend nights.

This feature provides additional information about the distribution of the customer's stay.

# 9. Aggregations

Aggregations summarize multiple numerical values into a single statistic, such as sum, mean, minimum, maximum, or count.

**When to use:** when multiple records belong to the same entity and we need a summary of those records, such as summarizing hotel bookings by hotel type.

**Why:** aggregation converts detailed booking-level information into meaningful summary features that can help a Machine Learning model understand overall booking patterns.

In the Hotel Booking Demand dataset, we can calculate the average ADR and total stay nights for each hotel type.

```python
# Create hotel-level aggregation features

hotel_features = df.groupby("hotel").agg(
    Average_ADR=("adr", "mean"),
    Average_Stay_Nights=("total_stay_nights", "mean"),
    Total_Bookings=("hotel", "count")
)

hotel_features
```

### Observation

The aggregation features summarize booking information at the hotel level.

- `Average_ADR` represents the average daily rate for each hotel type.
- `Average_Stay_Nights` represents the average length of stay.
- `Total_Bookings` represents the number of bookings for each hotel type.

These aggregated features provide a higher-level view of the booking data and can be useful for understanding differences between hotel types.

# 10. Log Transformation

Log transformation compresses large numerical values and reduces the effect of extreme values and right-skewed distributions.

**When to use:** when a numerical feature is heavily right-skewed and contains large values, such as `lead_time` in the Hotel Booking Demand dataset.

**Why:** log transformation reduces the influence of very large values and can make a skewed numerical feature more suitable for Machine Learning models.

In the Hotel Booking Demand dataset, `lead_time` represents the number of days between booking and arrival. It can contain small as well as very large values, so a log transformation can help reduce its skewness.

```python
# Apply log transformation to lead_time

df["lead_time_log"] = np.log1p(df["lead_time"])

df[["lead_time", "lead_time_log"]].describe()
```

### Observation

The `lead_time_log` feature is the log-transformed version of `lead_time`.

The transformation compresses large values while retaining the relative ordering of the observations.

This can reduce the impact of extreme values and make the feature more suitable for further Machine Learning analysis.

# 11. Square Root Transformation

Square root transformation reduces skewness in numerical features with a gentler effect than log transformation.

**When to use:** when a numerical feature is moderately right-skewed and contains zero values.

**Why:** it reduces the influence of large values while preserving more of the original variation compared with a log transformation.

In the Hotel Booking Demand dataset, `stays_in_week_nights` represents the number of weekday nights in a booking. A square root transformation can be used to reduce the influence of bookings with very long weekday stays.

```python
# Apply square root transformation to stays_in_week_nights

df["week_nights_sqrt"] = np.sqrt(
    df["stays_in_week_nights"].clip(lower=0)
)

df[
    ["stays_in_week_nights", "week_nights_sqrt"]
].describe()
```

### Observation

The `week_nights_sqrt` feature is the square root transformed version of `stays_in_week_nights`.

The transformation reduces the effect of larger values while keeping zero values valid.

This can be useful when a numerical feature is moderately skewed and a gentler transformation is preferred.

# 12. Power Transformation

Power transformation changes the distribution of a numerical feature by applying a mathematical power transformation.

**When to use:** when the relationship between a numerical feature and the target is non-linear, or when the original feature distribution is highly skewed.

**Why:** power transformations can reduce skewness, stabilize variance, and make numerical features more suitable for Machine Learning models.

In the Hotel Booking Demand dataset, `adr` represents the Average Daily Rate. A Yeo-Johnson Power Transformation can be used because it can handle zero and negative values.

```python
from sklearn.preprocessing import PowerTransformer

# Apply Yeo-Johnson power transformation to adr

pt = PowerTransformer(method="yeo-johnson")

df["adr_power"] = pt.fit_transform(
    df[["adr"]]
)

df[
    ["adr", "adr_power"]
].describe()
```

### Observation

The `adr_power` feature is the power-transformed version of `adr`.

The Yeo-Johnson transformation helps reduce skewness and can make the distribution of the numerical feature more suitable for Machine Learning.

The transformed feature can be useful when the relationship between the original feature and the target is non-linear or when the feature has a highly skewed distribution.

# 13. Absolute Difference

The absolute difference measures the magnitude of the difference between two numerical values without considering the direction of the difference.

**When to use:** when only the size of the difference matters, not whether one value is higher or lower than the other.

**Why:** it prevents positive and negative differences from cancelling each other out and provides a direct measure of how far one value is from another.

In the Hotel Booking Demand dataset, we can calculate the absolute difference between the weekday stay nights and weekend stay nights.

```python
# Calculate absolute difference between weekday and weekend nights

df["abs_stay_nights_difference"] = (
    df["stays_in_week_nights"] -
    df["stays_in_weekend_nights"]
).abs()

df[
    [
        "stays_in_week_nights",
        "stays_in_weekend_nights",
        "abs_stay_nights_difference"
    ]
].head()
```

### Observation

The `abs_stay_nights_difference` feature represents the magnitude of the difference between weekday and weekend stay nights.

The absolute value removes the direction of the difference, so the feature always represents how large the difference is.

# 14. Relative Difference

Relative difference expresses the difference between two related numerical values as a proportion of a reference value.

**When to use:** when the same absolute difference can have different meanings depending on the scale of the values being compared.

**Why:** it allows fair comparison between values with different scales by expressing the difference relative to a reference value.

In the Hotel Booking Demand dataset, we can calculate the relative difference between weekday and weekend stay nights using the total stay nights as the reference value.

```python
# Calculate relative difference between weekday and weekend stay nights

df["relative_stay_difference"] = (
    (
        df["stays_in_week_nights"] -
        df["stays_in_weekend_nights"]
    ) /
    df["total_stay_nights"].replace(0, 1)
)

df[
    [
        "stays_in_week_nights",
        "stays_in_weekend_nights",
        "total_stay_nights",
        "relative_stay_difference"
    ]
].head()
```

### Observation

The `relative_stay_difference` feature represents the difference between weekday and weekend stay nights relative to the total stay duration.

This provides a scale-independent measure that makes it easier to compare bookings with different total stay durations.

# 15. Rate Features

A rate feature expresses a quantity per unit of time or per unit of another relevant dimension.

**When to use:** when comparing bookings with different stay durations or different levels of activity, where the raw total may not provide a fair comparison.

**Why:** rate features normalize a quantity based on a relevant unit, making comparisons between different bookings easier.

In the Hotel Booking Demand dataset, we can calculate the number of special requests per stay night.

```python
# Calculate special requests per stay night

df["special_requests_per_night"] = (
    df["total_of_special_requests"] /
    df["total_stay_nights"].replace(0, 1)
)

df[
    [
        "total_of_special_requests",
        "total_stay_nights",
        "special_requests_per_night"
    ]
].head()
```

### Observation

The `special_requests_per_night` feature represents the number of special requests relative to the total number of stay nights.

This rate provides a normalized measure that makes it easier to compare bookings with different stay durations.

# 16. Normalized Features

Normalization rescales a numerical feature into a fixed range, typically between 0 and 1, based on its minimum and maximum values.

**When to use:** when numerical features have very different ranges and the Machine Learning algorithm is sensitive to feature magnitude, such as distance-based or gradient-based models.

**Why:** normalization prevents features with larger numerical ranges from dominating the model simply because of their scale.

In the Hotel Booking Demand dataset, `lead_time` and `adr` have different numerical ranges. We can normalize them to a common range between 0 and 1.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[
    ["lead_time_norm", "adr_norm"]
] = scaler.fit_transform(
    df[["lead_time", "adr"]]
)

df[
    [
        "lead_time",
        "lead_time_norm",
        "adr",
        "adr_norm"
    ]
].head()
```

### Observation

The `lead_time` and `adr` features were transformed into normalized features between 0 and 1.

Normalization puts numerical features on a common scale, which can be useful for Machine Learning algorithms that are sensitive to feature magnitude.

In a real Machine Learning workflow, the scaler should be fitted only on the training data and then applied to validation and test data to prevent data leakage.

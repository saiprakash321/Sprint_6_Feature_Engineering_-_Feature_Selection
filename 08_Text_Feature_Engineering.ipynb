# Module 8 - Text Feature Engineering

## 1. Text Length

Definition: Text length is the total number of characters present in a piece of text, including spaces.

Example: "Hotel Booking" has a length of 13.

Why it is used?

Text length is a simple feature that can describe the size of text and may help a Machine Learning model identify patterns.

Hotel Example:

The `reservation_status` column contains text values such as "Check-Out", "Canceled", and "No-Show".

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# Handle missing values and calculate text length

df["reservation_status"] = (
    df["reservation_status"]
    .fillna("")
    .astype(str)
)

df["Text_Length"] = df["reservation_status"].apply(len)

df[
    ["reservation_status", "Text_Length"]
].head()
```

### Observation

The `Text_Length` feature contains the number of characters in each `reservation_status` value.

This converts a text property into a numerical feature that can be used by a Machine Learning model.

## 2. Character Count

Character Count represents the total number of characters present in a text value.

It is similar to Text Length, but it is useful when we specifically want to measure the size of textual content.

### Example

Text: `"Hotel Booking"`

Character Count: `13`

### Why Character Count is Useful

Character count can help identify:

- Short or long text values
- Differences in text size
- Unusual or inconsistent text entries
- Patterns that may be useful for machine learning

### Hotel Booking Example

In the Hotel Booking dataset, `reservation_status` contains text values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

We can calculate the number of characters in each value.

```python
df["Character_Count"] = df["reservation_status"].apply(len)

df[["reservation_status", "Character_Count"]].head()
```

### Observation

The `Character_Count` feature represents the number of characters present in each `reservation_status` value.

Different text values can have different character counts.

## 3. Word Count

Word Count represents the total number of words present in a text value.

It helps us understand how much textual information is contained in a particular value.

### Example

Text: `"Hotel Booking"`

Word Count: `2`

### Why Word Count is Useful

Word count can help identify:

- Short or long text values
- Differences in the amount of textual information
- Text patterns that may be useful for machine learning
- Unusual text entries

### Hotel Booking Example

In the Hotel Booking dataset, `reservation_status` contains text values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

We can calculate the number of words in each value.

```python
df["Word_Count"] = df["reservation_status"].apply(
    lambda x: len(x.split())
)

df[["reservation_status", "Word_Count"]].head()
```

### Observation

The `Word_Count` feature represents the number of words present in each `reservation_status` value.

It converts the text information into a numerical feature that can be used for analysis and machine learning.

## 4. Sentence Count

Sentence Count represents the total number of sentences present in a text value.

It helps identify how much sentence-level information is contained in a text.

### Example

Text: `"Hotel booking is confirmed. Customer requested a room."`

Sentence Count: `2`

### Why Sentence Count is Useful

Sentence count can help identify:

- Short or long text content
- The amount of information in a text value
- Differences between text entries
- Text patterns that may be useful for machine learning

### Hotel Booking Example

The Hotel Booking dataset contains text values in the `reservation_status` column.

We can count the sentences in each value by checking sentence-ending punctuation such as `.`, `!`, and `?`.

```python
df["Sentence_Count"] = df["reservation_status"].apply(
    lambda x: sum(x.count(p) for p in [".", "!", "?"])
)

df[["reservation_status", "Sentence_Count"]].head()
```

## 5. Number of Digits

Number of Digits represents the total number of numerical digits present in a text value.

It helps identify whether text contains numbers and how many numerical characters are included.

### Example

Text: `"Room 101"`

Number of Digits: `3`

### Why Number of Digits is Useful

Number of digits can help identify:

- Numerical information present in text
- Booking or reference numbers
- Dates and other numeric patterns
- Text patterns that may be useful for machine learning

### Hotel Booking Example

The `reservation_status` column contains text values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

We can count the numerical digits present in each value.

```python
df["Digit_Count"] = df["reservation_status"].apply(
    lambda x: sum(char.isdigit() for char in x)
)

df[["reservation_status", "Digit_Count"]].head()
```

## 6. Number of Special Characters

Number of Special Characters represents the total number of characters in a text that are not letters or numbers.

Examples of special characters include:

- `@`
- `#`
- `$`
- `%`
- `-`
- `!`

### Example

Text: `"Hotel@2026!"`

Number of Special Characters: `2`

### Why Special Character Count is Useful

Special character count can help identify:

- Formatting patterns in text
- Symbols present in text
- Unusual or inconsistent text values
- Text patterns that may be useful for machine learning

### Hotel Booking Example

The `reservation_status` column contains values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

The hyphen (`-`) is a special character, so we can count such characters in each value.

```python
df["Special_Character_Count"] = df["reservation_status"].apply(
    lambda x: sum(not char.isalnum() and not char.isspace() for char in x)
)

df[["reservation_status", "Special_Character_Count"]].head()
```

## 7. Uppercase Count

Uppercase Count represents the total number of uppercase letters present in a text value.

It helps identify capitalization patterns in textual data.

### Example

Text: `"HOTEL Booking"`

Uppercase Count: `7`

### Why Uppercase Count is Useful

Uppercase count can help identify:

- Capitalization patterns
- Formatting differences
- Text inconsistencies
- Text patterns that may be useful for machine learning

### Hotel Booking Example

The `reservation_status` column contains values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

We can count the uppercase letters present in each value.

```python
df["Uppercase_Count"] = df["reservation_status"].apply(
    lambda x: sum(char.isupper() for char in x)
)

df[["reservation_status", "Uppercase_Count"]].head()
```

## 8. Lowercase Count

Lowercase Count represents the total number of lowercase letters present in a text value.

It helps identify the capitalization pattern of textual data.

### Example

Text: `"Hotel Booking"`

Lowercase Count: `11`

### Why Lowercase Count is Useful

Lowercase count can help identify:

- Capitalization patterns
- Formatting differences
- Text inconsistencies
- Text patterns that may be useful for machine learning

### Hotel Booking Example

The `reservation_status` column contains values such as:

- `Check-Out`
- `Canceled`
- `No-Show`

We can count the lowercase letters present in each value.

```python
df["Lowercase_Count"] = df["reservation_status"].apply(
    lambda x: sum(char.islower() for char in x)
)

df[["reservation_status", "Lowercase_Count"]].head()
```

## 9. Keyword Count

Keyword Count represents the number of times a specific keyword appears in a text value.

It helps identify whether important words or terms are present in textual data.

### Example

Text: `"Hotel booking confirmed"`

Keyword: `"booking"`

Keyword Count: `1`

### Why Keyword Count is Useful

Keyword count can help identify:

- Important words in text
- Relevant categories or topics
- Specific patterns in textual data
- Features that may be useful for machine learning

### Hotel Booking Example

In the Hotel Booking dataset, we can check whether specific keywords such as `"canceled"` appear in the `reservation_status` values.

We can count the occurrence of the keyword in each text value.

```python
keyword = "canceled"

df["Keyword_Count"] = df["reservation_status"].str.lower().apply(
    lambda x: x.count(keyword)
)

df[["reservation_status", "Keyword_Count"]].head()
```

## 10. Bag of Words

Bag of Words (BoW) is a text feature engineering technique that converts text into numerical features based on the words present in the text.

It creates a vocabulary from the text and represents each text value using the frequency of those words.

### Example

Text:

`"hotel booking"`

Vocabulary:

`["hotel", "booking"]`

The text can be represented numerically based on the occurrence of these words.

### Why Bag of Words is Useful

Bag of Words can help:

- Convert text into numerical features
- Identify important words
- Represent text for machine learning models
- Analyse word frequency

### Hotel Booking Example

For the Hotel Booking dataset, we can apply Bag of Words to the `reservation_status` column.

The text values are converted into numerical vectors based on the words present in the dataset.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

bow_features = vectorizer.fit_transform(
    df["reservation_status"].fillna("").astype(str)
)

bow_df = pd.DataFrame(
    bow_features.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_df.head()
```

## 11. TF-IDF

TF-IDF stands for Term Frequency-Inverse Document Frequency.

It is a text feature engineering technique that converts text into numerical features based on the importance of words.

TF-IDF gives higher importance to words that are frequent in a particular text but less common across the complete collection of texts.

### Example

If a word appears frequently in one text but rarely in other texts, it receives a higher TF-IDF value.

### Why TF-IDF is Useful

TF-IDF can help:

- Convert text into numerical features
- Identify important words
- Reduce the importance of very common words
- Represent text for machine learning models

### Hotel Booking Example

For the Hotel Booking dataset, TF-IDF can be applied to the `reservation_status` column to represent the text values numerically.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_features = tfidf_vectorizer.fit_transform(
    df["reservation_status"].fillna("").astype(str)
)

tfidf_df = pd.DataFrame(
    tfidf_features.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

tfidf_df.head()
```

## 12. N-Grams

N-Grams are combinations of consecutive words in a text.

They help capture relationships between words instead of considering each word independently.

### Types of N-Grams

- Unigram: One word
- Bigram: Two consecutive words
- Trigram: Three consecutive words

### Example

Text:

`"hotel booking confirmed"`

Unigrams:
`hotel`, `booking`, `confirmed`

Bigrams:
`hotel booking`, `booking confirmed`

Trigrams:
`hotel booking confirmed`

### Why N-Grams are Useful

N-Grams can help:

- Capture relationships between words
- Identify common word combinations
- Preserve some word-order information
- Create useful text features for machine learning

### Hotel Booking Example

We can create N-Gram features from the `reservation_status` column.

Since the status values are short, bigrams may be limited, but the technique can still be demonstrated.

```python
from sklearn.feature_extraction.text import CountVectorizer

ngram_vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

ngram_features = ngram_vectorizer.fit_transform(
    df["reservation_status"].fillna("").astype(str)
)

ngram_df = pd.DataFrame(
    ngram_features.toarray(),
    columns=ngram_vectorizer.get_feature_names_out()
)

ngram_df.head()
```

## How Raw Text Becomes Machine Learning Features

Raw text cannot usually be directly used by most machine learning models.

Text Feature Engineering converts raw text into meaningful numerical features.

### Structural Text Features

These features describe the structure of the text:

- Text Length
- Character Count
- Word Count
- Sentence Count
- Digit Count
- Special Character Count
- Uppercase Count
- Lowercase Count

### Content-Based Text Features

These features represent the actual content and word patterns:

- Keyword Count
- Bag of Words
- TF-IDF
- N-Grams

### Overall Process

Raw Text
↓
Text Cleaning
↓
Text Analysis
↓
Text Feature Engineering
↓
Numerical Features
↓
Machine Learning Model

These numerical features can then be used as inputs to machine learning algorithms.

### Observation

Text Feature Engineering converts textual information into numerical features that machine learning models can understand.

Structural features describe the properties of the text, while content-based techniques such as Bag of Words, TF-IDF, and N-Grams represent the words and patterns within the text.

The notebook demonstrated different techniques for converting text into numerical features, including text statistics, keyword counts, Bag of Words, TF-IDF, and N-Grams.

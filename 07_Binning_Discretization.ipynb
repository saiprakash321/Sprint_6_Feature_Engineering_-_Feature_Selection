# Module 7 - Binning & Discretization

## 1. What is Binning?

Definition: Binning means grouping continuous numerical values into a smaller number of ranges called bins.

Example:

An ADR value of 45 can be grouped into a "Low" or "Medium" price range instead of using the exact value 45.

Hotel Example:

ADR → Low ADR, Medium ADR, High ADR

Why it is used?

Binning makes numerical data easier to understand, reduces small variations in values, and can help identify useful patterns in different groups.

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# View ADR values before binning

df["adr"].head()
```

### Observation

The ADR column contains continuous numerical values representing the average daily rate of a hotel booking.

Binning can be used to convert these continuous values into meaningful groups such as Low, Medium, and High ADR.

## 2. Why Binning?

Definition: Binning is used when continuous numerical values are too detailed, spread out, or difficult to interpret directly.

Example:

ADR values such as 45.20, 45.50, and 46.10 can be grouped into a common price category.

Hotel Example:

ADR → Low ADR, Medium ADR, High ADR

Why it is used?

Binning can make numerical data easier to understand, reduce the effect of small variations, and help identify patterns across different groups.

It can also make data easier to visualize and use in business analysis.

```python
# Check the distribution of ADR

df["adr"].describe()
```

### Observation

The `adr` column contains a wide range of numerical values.

Viewing the minimum, maximum, mean, and quartiles helps us understand the distribution before applying binning.

## 3. Equal-Width Binning

Definition: Equal-width binning divides the complete range of numerical values into bins having the same width.

Example:

If ADR values range from 0 to 100 and we create 5 bins, each bin will have a width of 20.

Hotel Example:

ADR values can be divided into 5 equal-width ranges.

Why it is used?

It is a simple and quick method for grouping continuous numerical data into equally sized ranges.

However, if the data is unevenly distributed, some bins may contain many values while others may contain very few values.

```python
# Create 5 equal-width bins for ADR

df["ADR_EqualWidth"] = pd.cut(
    df["adr"],
    bins=5
)

df["ADR_EqualWidth"].value_counts()
```

### Observation

The `adr` values have been divided into 5 equal-width ranges.

The number of bookings in each bin may be different because equal-width binning focuses on the range of values, not on distributing the same number of records into each bin.

## 4. Equal-Frequency Binning

Definition: Equal-frequency binning divides the data so that each bin contains approximately the same number of records.

Example:

If there are 100 records and we create 4 bins, each bin will contain approximately 25 records, even if the value ranges are different.

Hotel Example:

ADR values can be divided into 4 groups based on the number of bookings in each group.

Why it is used?

It is useful when the data is unevenly distributed because it prevents most records from being concentrated in only one or two bins.

```python
# Create 4 equal-frequency bins for ADR

df["ADR_EqualFreq"] = pd.qcut(
    df["adr"],
    q=4,
    duplicates="drop"
)

df["ADR_EqualFreq"].value_counts()
```

### Observation

The `adr` values have been divided into approximately equal-sized groups based on the number of booking records.

Unlike equal-width binning, the ranges of the bins are not necessarily the same.

## 5. Quantile Binning

Definition: Quantile binning divides numerical data using percentile values so that each group contains approximately the same number of records.

Example:

A numerical feature can be divided using the 25th, 50th, and 75th percentiles to create Low, Medium, High, and Very High groups.

Hotel Example:

ADR can be grouped into Low, Medium, High, and Very High based on its position in the dataset.

Why it is used?

Quantile binning is useful when numerical values are unevenly distributed. It creates groups based on the relative position of each value in the dataset.

```python
# Create quantile-based ADR groups

df["ADR_Quantile"] = pd.qcut(
    df["adr"],
    q=[0, 0.25, 0.50, 0.75, 1],
    labels=["Low", "Medium", "High", "Very High"],
    duplicates="drop"
)

df["ADR_Quantile"].value_counts()
```

### Observation

The ADR values have been divided into four groups based on their position in the dataset.

The groups represent relatively Low, Medium, High, and Very High ADR values.

## 6. Domain-Based Binning

Definition: Domain-based binning creates groups using real-world knowledge, business rules, or meaningful thresholds rather than statistical methods.

Example:

Age can be grouped into Child, Teen, Adult, and Senior based on commonly used age ranges.

Hotel Example:

ADR can be grouped into Low, Medium, and High price categories using meaningful business-defined ranges.

Why it is used?

Domain-based bins are easier to understand and explain because the boundaries have a clear real-world meaning.

When to use:

Use domain-based binning when business knowledge provides meaningful thresholds for a numerical feature.

```python
# Create domain-based ADR categories

bins = [-np.inf, 50, 100, np.inf]
labels = ["Low ADR", "Medium ADR", "High ADR"]

df["ADR_Domain_Group"] = pd.cut(
    df["adr"],
    bins=bins,
    labels=labels
)

df[
    ["adr", "ADR_Domain_Group"]
].head()
```

### Observation

The ADR values have been grouped into Low ADR, Medium ADR, and High ADR categories using fixed domain-based thresholds.

These categories are easier to interpret than raw numerical ADR values.

## 7. Continuous → Categorical

Definition: Continuous to categorical conversion means transforming numerical values into a smaller number of meaningful categories using binning.

Example:

Age values → Young, Middle-Aged, Senior

Hotel Example:

ADR values → Low ADR, Medium ADR, High ADR

Why it is used?

Converting continuous values into categories makes the data easier to understand, group, filter, and visualize.

When to use:

Use this approach when ranges or categories are more meaningful for analysis than the exact numerical values.

A limitation is that converting continuous values into categories can result in some loss of detailed information.

```python
# Convert continuous ADR values into categories

bins = [-np.inf, 50, 100, np.inf]
labels = ["Low", "Medium", "High"]

df["ADR_Category"] = pd.cut(
    df["adr"],
    bins=bins,
    labels=labels
)

df[
    ["adr", "ADR_Category"]
].head()
```

### Observation

The continuous `adr` values have been converted into three categorical groups: Low, Medium, and High.

This makes the numerical feature easier to interpret and use for grouped analysis.

## 9. Income Groups

Definition: Income group binning divides income or spending values into categories such as Low, Medium, and High.

Example:

Low Income → Medium Income → High Income

Hotel Example:

The Hotel Booking dataset does not contain an income column. Therefore, income groups cannot be created directly from the available data.

As an alternative, ADR can be grouped into price levels because ADR represents the daily room rate.

Why it is used?

Income or spending groups can help identify different customer segments and support targeted business decisions.

```python
# Use ADR as a domain-related alternative to income

df["ADR_Price_Group"] = pd.qcut(
    df["adr"],
    q=3,
    labels=["Low", "Medium", "High"],
    duplicates="drop"
)

df[
    ["adr", "ADR_Price_Group"]
].head()
```

### Observation

The dataset does not contain customer income information.

Therefore, ADR was used as an alternative numerical feature to demonstrate grouping values into Low, Medium, and High categories.

## 10. Risk Categories

Definition: Risk category binning converts a numerical risk score or related numerical value into meaningful risk levels.

Example:

0–30 → Low Risk  
31–60 → Medium Risk  
61–100 → High Risk

Hotel Example:

The Hotel Booking dataset does not contain a predefined risk score.

As a practical example, lead time can be grouped into Low, Medium, and High categories because longer lead times can be useful for analyzing booking patterns.

Why it is used?

Risk categories make numerical values easier to interpret and help identify groups that may require different attention or analysis.

```python
# Create lead-time risk categories

bins = [-np.inf, 30, 90, np.inf]
labels = ["Low", "Medium", "High"]

df["LeadTime_Risk_Category"] = pd.cut(
    df["lead_time"],
    bins=bins,
    labels=labels
)

df[
    ["lead_time", "LeadTime_Risk_Category"]
].head()
```

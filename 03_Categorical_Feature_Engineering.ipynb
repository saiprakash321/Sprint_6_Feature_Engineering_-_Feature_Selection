## 1. Categorical Variables

A categorical variable represents a limited and fixed set of categories or groups.

Categorical variables are used when a column contains labels or categories rather than continuous numerical values.

In the Hotel Booking dataset, examples include:

- `hotel` – City Hotel, Resort Hotel
- `meal` – BB, HB, FB, SC, Undefined
- `market_segment` – Online TA, Offline TA/TO, Groups, Direct, etc.
- `deposit_type` – No Deposit, Non Refund, Refundable
- `customer_type` – Contract, Group, Transient, Transient-Party

### When Categorical Variables Are Used

Categorical variables are used when:

- A column represents a group or category.
- The values are labels rather than measurements.
- The number of possible values is limited or belongs to a defined set.

### Why Categorical Variables Are Important

Most Machine Learning algorithms require numerical input.

Therefore, categorical variables usually need to be converted into numerical representations before they are provided to a Machine Learning model.

Proper encoding helps the model understand category information without introducing incorrect relationships between categories.

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Sprint_6_Feature_Engineering_-_Feature_Selection\data\hotel_bookings.csv"
)

df.head()
```

```python
# Display categorical columns and their unique values

categorical_columns = [
    "hotel",
    "meal",
    "market_segment",
    "deposit_type",
    "customer_type"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].dropna().unique())
```

### Observation

The selected columns contain a fixed set of categorical values representing different hotel booking characteristics.

These variables provide useful information about the booking and customer, but they need suitable encoding before being used in most Machine Learning models.

## 2. Nominal Variables

Nominal variables are categorical variables where the categories have no inherent order or ranking.

For example, in the Hotel Booking dataset:

- `meal` – BB, HB, FB, SC, Undefined
- `market_segment` – Online TA, Offline TA/TO, Groups, Direct, etc.
- `deposit_type` – No Deposit, Non Refund, Refundable
- `customer_type` – Contract, Group, Transient, Transient-Party

There is no meaningful order between these categories.

For example, `Online TA` is not greater than `Direct`, and `No Deposit` is not lower than `Refundable`.

### When Nominal Variables Are Used

Nominal variables are used when:

- Categories represent different groups.
- Categories do not have a meaningful ranking.
- The values are labels rather than ordered levels.

### Why Nominal Variables Are Important

Nominal variables provide useful information about the characteristics of each booking.

However, Machine Learning models generally cannot directly process text categories. Therefore, nominal variables are usually converted into numerical representations using techniques such as One-Hot Encoding.

One-Hot Encoding is commonly suitable for nominal variables because it does not introduce a false ordering between categories.

```python
# Check nominal categorical variables

nominal_columns = [
    "meal",
    "market_segment",
    "deposit_type",
    "customer_type"
]

for column in nominal_columns:
    print(f"\n{column}")
    print(df[column].value_counts().head())
```

### Observation

The selected columns contain nominal categories with no inherent order.

For example, different `market_segment` values represent different booking channels, but one category is not numerically higher or lower than another.

Therefore, these variables can be encoded using techniques such as One-Hot Encoding without creating an artificial ranking between categories.

## 3. Ordinal Variables

Ordinal variables are categorical variables where the categories have a meaningful order or ranking.

Unlike nominal variables, ordinal categories have a natural sequence.

For example:

- Short < Medium < Long
- Low < Medium < High
- Beginner < Intermediate < Advanced

The difference between the categories does not necessarily have to be equal. The important point is that the order has meaning.

### Hotel Booking Example

The Hotel Booking dataset does not contain a direct ordinal categorical column. Therefore, an ordinal feature can be created from an existing numerical feature such as `lead_time`.

For example:

- Short Lead Time – 0 to 30 days
- Medium Lead Time – 31 to 90 days
- Long Lead Time – more than 90 days

These categories have a meaningful order because a longer lead time represents a booking made further in advance.

### When Ordinal Variables Are Used

Ordinal variables are used when:

- Categories have a meaningful ranking.
- The order of categories contains useful information.
- The categories represent levels or stages.

### Why Ordinal Variables Are Important

Ordinal variables allow Machine Learning models to use the ranking information contained in categories.

They can be converted into numerical values while preserving their order.

```python
# Create an ordinal feature from lead_time

df["lead_time_category"] = pd.cut(
    df["lead_time"],
    bins=[-1, 30, 90, np.inf],
    labels=["Short", "Medium", "Long"]
)

df[["lead_time", "lead_time_category"]].head()
```

### Observation

The `lead_time_category` feature contains three ordered categories:

- Short
- Medium
- Long

The categories have a meaningful ranking based on the number of days between booking and arrival.

Since the order contains useful information, this feature can be treated as an ordinal variable and later encoded using an ordinal encoding technique.

## 4. Binary Variables

Binary variables are categorical variables that contain exactly two possible categories or values.

In the Hotel Booking dataset, `is_repeated_guest` is a binary variable.

Its values are:

- 0 – Not a repeated guest
- 1 – Repeated guest

Binary variables are often represented directly using 0 and 1.

### When Binary Variables Are Used

Binary variables are used when:

- A feature has exactly two possible outcomes.
- The categories represent Yes/No conditions.
- The values can naturally be represented as 0 and 1.

### Why Binary Variables Are Important

Binary variables are simple and efficient for Machine Learning models because they already have a numerical representation.

No additional encoding is required when the two categories are correctly represented as 0 and 1.

```python
# Check the binary variable

print(df["is_repeated_guest"].value_counts())

print("\nUnique values:")
print(df["is_repeated_guest"].unique())

df[["is_repeated_guest"]].head()
```

### Observation

The `is_repeated_guest` column contains two values: 0 and 1.

Here:

- 0 represents a guest who is not a repeated guest.
- 1 represents a repeated guest.

Since the feature is already represented using two numerical values, it can be directly used by Machine Learning models without additional categorical encoding.

## 5. One-Hot Encoding

One-Hot Encoding converts each category of a categorical variable into a separate binary column.

For example, if `deposit_type` contains:

- No Deposit
- Non Refund
- Refundable

One-Hot Encoding creates separate columns for each category.

Each row receives:

- 1 if the booking belongs to that category
- 0 otherwise

### When One-Hot Encoding Is Used

One-Hot Encoding is commonly used when:

- The categorical variable is nominal.
- Categories do not have a meaningful order.
- The number of categories is small or moderate.

### Why One-Hot Encoding Is Important

One-Hot Encoding allows Machine Learning models to use categorical information without introducing a false numerical ranking.

For example, assigning:

`No Deposit = 1`, `Non Refund = 2`, `Refundable = 3`

could incorrectly suggest that one category is greater than another.

One-Hot Encoding avoids this problem by creating independent binary columns for each category.

### Limitation

If a categorical variable has a very large number of unique categories, One-Hot Encoding can create many additional columns and increase the dimensionality of the dataset.

```python
# Apply One-Hot Encoding to a nominal variable

deposit_encoded = pd.get_dummies(
    df["deposit_type"],
    prefix="deposit_type"
)

deposit_encoded.head()
```

### Observation

The `deposit_type` column has been converted into separate binary columns.

Each booking is represented using 0 or 1 for each deposit category.

This preserves the categorical information without creating an artificial order between the categories.

One-Hot Encoding is therefore suitable for nominal variables such as `deposit_type`.

## 6. Ordinal Encoding

Ordinal Encoding converts categorical values into numerical values while preserving their meaningful order.

For example, the `lead_time_category` feature created earlier contains:

- Short
- Medium
- Long

These categories have a natural order:

Short < Medium < Long

They can therefore be represented as:

- Short → 0
- Medium → 1
- Long → 2

### When Ordinal Encoding Is Used

Ordinal Encoding is used when:

- Categories have a meaningful order.
- The ranking between categories contains useful information.
- The numerical values should preserve that ranking.

### Why Ordinal Encoding Is Important

Ordinal Encoding allows Machine Learning models to use the ordering information present in categorical variables.

It is more appropriate for ordinal variables than assigning arbitrary numerical values to nominal categories.

```python
# Apply Ordinal Encoding to the lead time categories

ordinal_mapping = {
    "Short": 0,
    "Medium": 1,
    "Long": 2
}

df["lead_time_category_encoded"] = (
    df["lead_time_category"]
    .map(ordinal_mapping)
)

df[
    [
        "lead_time_category",
        "lead_time_category_encoded"
    ]
].head()
```

### Observation

The `lead_time_category` values have been converted into numerical values while maintaining their order.

Short is represented by 0, Medium by 1, and Long by 2.

The encoding preserves the meaningful ranking between the categories, making it suitable for Machine Learning models.

## 7. Label Encoding

Label Encoding converts each categorical value into a unique integer.

For example, a categorical column such as `market_segment` may contain values such as:

- Online TA
- Offline TA/TO
- Direct
- Groups

Label Encoding assigns a different integer to each category.

For example:

- Direct → 0
- Groups → 1
- Offline TA/TO → 2
- Online TA → 3

The assigned numbers are labels and do not represent an actual ranking between the categories.

### When Label Encoding Is Used

Label Encoding can be used when:

- A categorical variable needs to be represented using integer values.
- The number of categories needs to be converted into a compact numerical representation.
- It is suitable for certain Machine Learning workflows, particularly some tree-based models.

For nominal variables, One-Hot Encoding is generally safer when practical because Label Encoding can introduce an artificial numerical relationship between categories.

### Why Label Encoding Is Important

Label Encoding provides a simple way to convert categorical values into numerical values.

However, the encoding method should be selected based on the type of categorical variable and the Machine Learning algorithm being used.

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy of the categorical column
label_encoder = LabelEncoder()

df["market_segment_encoded"] = label_encoder.fit_transform(
    df["market_segment"].fillna("Unknown")
)

df[
    [
        "market_segment",
        "market_segment_encoded"
    ]
].head()
```

### Observation

The `market_segment` categories have been converted into integer labels.

Each category receives a unique numerical value.

These numbers represent category labels and do not indicate that one market segment is higher or lower than another.

Therefore, Label Encoding should be used carefully for nominal variables, depending on the Machine Learning model.

## 8. Frequency Encoding

Frequency Encoding replaces each category with its frequency or proportion in the dataset.

For example, the `country` column contains many different countries. Instead of creating a separate column for every country, each country can be represented by how frequently it appears in the dataset.

For example:

- PRT → frequency of bookings from Portugal
- GBR → frequency of bookings from United Kingdom
- FRA → frequency of bookings from France

The frequency can be represented as either:

- Count of occurrences
- Proportion of total observations

### When Frequency Encoding Is Used

Frequency Encoding is useful when:

- A categorical variable has many unique categories.
- One-Hot Encoding would create too many columns.
- Category frequency contains useful information.

### Why Frequency Encoding Is Important

Frequency Encoding converts a high-cardinality categorical variable into a single numerical feature.

This reduces the number of columns while retaining information about how common each category is.

However, categories with the same frequency will receive the same encoded value, even if they are different categories.

```python
# Frequency Encoding for country

country_frequency = df["country"].value_counts(normalize=True)

df["country_frequency"] = (
    df["country"]
    .map(country_frequency)
    .fillna(0)
)

df[
    [
        "country",
        "country_frequency"
    ]
].head()
```

### Observation

The `country` column has been converted into a numerical frequency feature.

Each country is represented by the proportion of bookings associated with that country.

Frequency Encoding keeps the information in a single column, making it useful for high-cardinality categorical variables such as `country`.

## 9. Target Encoding

Target Encoding replaces each category with a numerical value calculated from the target variable.

In the Hotel Booking dataset, the target variable is `is_canceled`:

- 0 – Booking was not canceled
- 1 – Booking was canceled

For example, Target Encoding can calculate the average cancellation rate for each `market_segment`.

If a particular market segment has a cancellation rate of 0.30, that category can be represented by 0.30.

### When Target Encoding Is Used

Target Encoding is useful when:

- A categorical variable has many unique categories.
- The target variable contains useful information about the categories.
- One-Hot Encoding would create too many columns.

### Why Target Encoding Is Important

Target Encoding converts categorical variables into a single numerical feature based on their relationship with the target.

However, Target Encoding can cause overfitting and target leakage if the target information is calculated using the same data used to train the model.

Therefore, in a proper Machine Learning workflow, target statistics should be calculated using only the training data or within cross-validation folds.

### Hotel Booking Example

For `market_segment`, the average value of `is_canceled` can represent the cancellation rate for each market segment.

```python
# Target Encoding using the training data only

from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["is_canceled"]
)

# Calculate cancellation rate for each market segment using training data
target_mean = train_df.groupby("market_segment")["is_canceled"].mean()

# Map the training and test categories
train_df["market_segment_target_encoded"] = (
    train_df["market_segment"]
    .map(target_mean)
)

test_df["market_segment_target_encoded"] = (
    test_df["market_segment"]
    .map(target_mean)
)

# Use the overall training target mean for unseen categories
global_mean = train_df["is_canceled"].mean()

train_df["market_segment_target_encoded"] = (
    train_df["market_segment_target_encoded"]
    .fillna(global_mean)
)

test_df["market_segment_target_encoded"] = (
    test_df["market_segment_target_encoded"]
    .fillna(global_mean)
)

train_df[
    [
        "market_segment",
        "is_canceled",
        "market_segment_target_encoded"
    ]
].head()
```

### Observation

The `market_segment` categories have been converted into numerical values based on their cancellation rate.

The target statistics were calculated only from the training data and then applied to both training and test data.

This approach helps prevent target leakage and makes Target Encoding safer for Machine Learning workflows.

## 10. Rare Category Grouping

Rare Category Grouping combines categories that occur very infrequently into a single category such as `Other`.

In the Hotel Booking dataset, the `country` column contains many different countries. Some countries have only a small number of bookings.

Instead of keeping every very rare country as a separate category, these categories can be grouped into `Other`.

### When Rare Category Grouping Is Used

Rare Category Grouping is used when:

- A categorical variable contains many rare categories.
- Some categories have very few observations.
- Rare categories may cause overfitting.
- The number of categories needs to be reduced.

### Why Rare Category Grouping Is Important

Grouping rare categories helps reduce unnecessary complexity and makes categorical features more stable.

It can also reduce the number of categories that need to be encoded.

### Hotel Booking Example

For the `country` feature, countries with fewer than a selected number of bookings can be grouped into `Other`.

The threshold should be selected based on the dataset size and business context.

```python
# Group rare countries into "Other"

country_counts = df["country"].value_counts()

rare_countries = country_counts[country_counts < 100].index

df["country_grouped"] = (
    df["country"]
    .where(~df["country"].isin(rare_countries), "Other")
)

df["country_grouped"].value_counts().head(10)
```

### Observation

Countries with fewer than 100 bookings have been grouped into the `Other` category.

This reduces the number of rare categories in the `country` feature and can help prevent the model from learning unreliable patterns from categories with very few observations.

The threshold of 100 is an example and can be adjusted according to the dataset and business requirements.

## 11. Category Cardinality

Category Cardinality refers to the number of unique categories present in a categorical variable.

For example, in the Hotel Booking dataset:

- `hotel` has very low cardinality.
- `meal` has low cardinality.
- `market_segment` has low to moderate cardinality.
- `country` has high cardinality.

### When Category Cardinality Is Important

Cardinality should be considered when:

- Selecting an encoding technique.
- Deciding whether One-Hot Encoding is suitable.
- Working with categorical variables containing many unique values.

### Why Category Cardinality Is Important

Low-cardinality variables can usually be handled easily using One-Hot Encoding.

High-cardinality variables can create a large number of columns when One-Hot Encoding is applied.

For high-cardinality variables, techniques such as Frequency Encoding, Target Encoding, or other suitable approaches can be considered.

```python
# Check the number of unique categories

categorical_columns = [
    "hotel",
    "meal",
    "market_segment",
    "country",
    "reserved_room_type"
]

cardinality = df[categorical_columns].nunique()

print("Category Cardinality:")
print(cardinality)
```

### Observation

The number of unique categories varies across the selected categorical variables.

`hotel` has low cardinality, while `country` has relatively high cardinality.

Therefore, the encoding technique should be selected based on the cardinality of each categorical variable.

Low-cardinality variables can generally use One-Hot Encoding, while high-cardinality variables may require more compact encoding techniques.

## 12. Handling Unknown Categories

Unknown categories are categories that appear in new or unseen data but were not present when the Machine Learning model was trained.

For example, a new hotel booking may contain a `market_segment` value that was not available in the training data.

If the encoding process does not handle this situation, the model may produce an error when processing the new category.

### When Unknown Categories Are Used

Handling unknown categories is important when:

- New categories may appear in future data.
- Training and testing data may contain different categories.
- The model will receive real-world data after deployment.

### Why Handling Unknown Categories Is Important

A Machine Learning pipeline should be able to process previously unseen categories without failing.

Unknown categories can be mapped to a common category such as `Unknown` or `Other`.

This makes the feature engineering process more robust when new data is received.

```python
# Handle an unseen market segment

known_segments = set(
    df["market_segment"].dropna().unique()
)

new_data = pd.Series([
    "Online TA",
    "Offline TA/TO",
    "New Segment"
])

new_data_handled = new_data.where(
    new_data.isin(known_segments),
    "Unknown"
)

print(new_data_handled)
```

### Observation

The existing `market_segment` categories are retained, while the unseen `New Segment` category is replaced with `Unknown`.

Handling unknown categories prevents errors when the Machine Learning model receives a category that was not present during training.

In a production preprocessing pipeline, the encoder should also be configured to handle unknown categories automatically.

## 13. Combining Categories

Combining Categories means merging or combining categorical values to create a new categorical feature.

This can be useful when two categorical variables together provide more meaningful information than either variable individually.

### Hotel Booking Example

In the Hotel Booking dataset, `hotel` and `market_segment` describe different aspects of a booking.

They can be combined to create a new feature:

`hotel_market_segment`

For example:

- City Hotel + Online TA
- City Hotel + Direct
- Resort Hotel + Online TA
- Resort Hotel + Groups

This creates a combined category representing both the type of hotel and the booking market segment.

### When Combining Categories Is Used

Combining categories is used when:

- Two categorical variables have meaningful relationships.
- A combined category provides additional business information.
- Domain knowledge suggests that the combination may help the Machine Learning model.

### Why Combining Categories Is Important

Combining categories can create meaningful interactions between categorical variables.

However, combining too many categorical variables can create a large number of unique combinations and increase model complexity.

```python
# Combine hotel and market segment

df["hotel_market_segment"] = (
    df["hotel"] + "_" + df["market_segment"]
)

df[
    [
        "hotel",
        "market_segment",
        "hotel_market_segment"
    ]
].head()

print("\nCombined category counts:")
print(df["hotel_market_segment"].value_counts().head(10))
```

### Observation

The `hotel` and `market_segment` columns have been combined into a new categorical feature called `hotel_market_segment`.

The new feature represents the relationship between the type of hotel and the market segment used for the booking.

This combined feature can provide additional information to a Machine Learning model when the interaction between these categories is meaningful.
